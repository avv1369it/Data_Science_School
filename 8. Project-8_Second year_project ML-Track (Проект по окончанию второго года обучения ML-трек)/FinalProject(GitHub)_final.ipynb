### 1.1. Проблема <a id="item-one-one"></a>

#### <span style="color:blue">**Компания ритейл (заказчик)** - поставил задачу повысить прибыль от дополнительных продаж в интернет-магазине, не менее чем на 20%, за счет внедрения рекомендательной системы, которая должна выводить 3 товара на главной странице сайта (посетителя).


***Рекомендательные системы*** — это технологии, предназначенные для предоставления персонализированных предложений пользователям на основе анализа их предыдущих действий и предпочтений. Они широко используются в различных сферах, таких как электронная коммерция, онлайн-медиа, социальные сети и другие.

Основная цель ***рекомендательных систем*** заключается в повышении удовлетворенности клиентов и увеличении доходов компании путем предложения продуктов или услуг, которые наиболее соответствуют потребностям конкретного пользователя. Это достигается путем сбора данных о поведении пользователей, их интересах и взаимодействиях с контентом, и последующего использования этих данных для формирования персональных рекомендаций.

В контексте электронной коммерции, например, ***рекомендательная система*** может предложить дополнительные товары, которые часто покупают вместе с текущим товаром, или товары, которые могут быть интересны клиенту на основании его предыдущих покупок или просмотров. Таким образом, компания увеличивает вероятность того, что клиент совершит дополнительную покупку, что ведет к росту выручки и прибыли.

***Рекомендательные системы*** могут использовать различные алгоритмы и подходы для генерации рекомендаций, включая классические методы, такие как коллаборативная фильтрация, а также современные методы машинного обучения, такие как глубокое обучение и нейронные сети. Выбор подхода зависит от объема и типа данных, доступных для анализа, а также от сложности задачи и специфических требований бизнеса.

Внедрение ***рекомендательной системы*** требует тщательного планирования и разработки, начиная от определения целей и метрик оценки эффективности до выбора методов и технологий для реализации. Важно также учитывать вопросы безопасности данных и конфиденциальности, так как рекомендации должны быть основаны на надежных и анонимизированных данных.

Таким образом, внедрение ***рекомендательной системы*** представляет собой сложный и многогранный процесс, требующий комплексного подхода и интеграции современных технологий, но при правильной реализации он способен значительно улучшить пользовательский опыт и финансовые показатели компании.

````python
# Опишем функцию на проверку возможности конвертации в int32
def check_int32_compatibility(arr):
    # Проверка минимального и максимального значений
    min_val = arr.min()
    max_val = arr.max()
    
    if min_val >= np.iinfo(np.int32).min and max_val <= np.iinfo(np.int32).max:
        return True
    else:
        return False

if __name__ == "__main__":
    
    print(f'Данные в visitorid в ev можно преобразовать в int32: {check_int32_compatibility(ev["visitorid"])}')
    print(f'Данные в itemid в ev можно преобразовать в int32: {check_int32_compatibility(ev["itemid"])}')
    print(f'Данные в itemid в pr можно преобразовать в int32: {check_int32_compatibility(pr["itemid"])}')
    #print(f'Данные в property в pr можно преобразовать в int32: {check_int32_compatibility(pr["property"])}')
````

````python
# Опишем функцию на проверку возможности конвертации в float32
def check_float32_compatibility(arr):
    # Проверка минимального и максимального значений
    min_val = arr.min()
    max_val = arr.max()
    
    if min_val >= np.iinfo(np.float32).min and max_val <= np.iinfo(np.float32).max:
        return True
    else:
        return False

if __name__ == "__main__":
    
    print(f'Данные в transactionid в ev можно преобразовать в float32: {check_int32_compatibility(ev["transactionid"])}')
    print(f'Данные в parentid в pr можно преобразовать в float32: {check_int32_compatibility(pr["parentid"])}')
````

#### 3.2.2 Создание дополнительных признаков (в том числе с помощью внешних источников) <a id="item-three-two-two"></a>

````python
# Создадим новый признак описывающий период времени в течение суток - указано в задании неделя 3
# Описываем функцию
def get_time_periods(hour):
    """Функция служит для создания нового признака - период времени в течение суток

    Args:
        hour (int): час в котором было совершено действие клиентом (просмотр, добавление в избранное, покупка)
    Returns:
        value (str): период времени в течение суток (Dawn / Рассвет, Morning / Утро, Afternoon / День, Evening / Вечер, Night / Ночь)
    """
    if hour >= 3 and hour < 7:
        return "Dawn"
    elif hour >= 7 and hour < 12:
        return "Morning"
    elif hour >= 12 and hour < 16:
        return "Afternoon"
    elif hour >= 16 and hour < 22:
        return "Evening"
    else:
        return "Night"
````

````python
# Описываем функцию вывода признаков с высокой кореляцией
def get_high_correlation(df, threshold=0.7):
    """
    Функция для получения пар признаков с высокой корреляцией в DataFrame.

    Args:
    - df (dataframe): исходный DataFrame.
    - threshold (float, optional): минимальный порог абсолютной величины коэффициента корреляции между двумя столбцами. По умолчанию = 0.7.

    Returns:
    - high_corr (dataframe): DataFrame, содержащий пары признаков с высокой корреляцие и само значение корреляции.
    """
    # Вычисляем матрицу корреляций
    corr = df.corr().abs()
    
    # Фильтруем матрицу, оставляя только элементы выше заданного порога
    high_corr = (corr[(corr > threshold) & (corr != 1)]
                .stack()
                .reset_index()
                .rename(columns={0: "correlation", "level_0": "feature_1", "level_1": "feature_2"}))
    
    # Убираем пары, где feature_1 == feature_2
    high_corr = high_corr[high_corr["feature_1"] != high_corr["feature_2"]]
    
    # Сортируем результат по убыванию корреляции
    high_corr = high_corr.sort_values(by="correlation", ascending=False)
    
    return high_corr
````

````python
# Опишем функцию для вывода отсортированных данных по заданному признаку
def highlight_feature(df, feature_name, num_obs=3):
    """
    Функция для создания стилизованного представления DataFrame, выделяющего строки с максимальными значениями по заданному признаку.

    Args:
    - df (dataframe): исходный DataFrame
    - feature_name (str): название столбца, по которому происходит сортировка
    - num_obs (int, optional): количество строк для отображения. По умолчанию = 3

    Returns:
    - styled_df (dataframe): стилизованный DataFrame с первыми num_obs строками, выделенными по заданному критерию
    """
    
    # Сортируем DataFrame по значению выбранного столбца в порядке убывания
    sorted_df = df.sort_values(by=feature_name, ascending=False)
    
    # Определяем функцию для выделения строк
    def highlight(s):
        if s.name == feature_name:
            return ['background-color: yellow']*len(s)
        else:
            return ['']*len(s)
    
    # Применяем функцию highlight к первым num_obs строкам и создаем стилизованный DataFrame
    styled_df = sorted_df.head(num_obs).style.apply(highlight)
    return styled_df
````

## 4. Создание факторов для моделирования <a id="item-four"></a>

Создание факторов для модели рекомендательной системы является ключевым шагом в разработке эффективной системы. Факторы представляют собой признаки или характеристики, которые используются для оценки предпочтений пользователей и свойств элементов. Генерация подходящих факторов позволяет модели рекомендательной системы предсказывать предпочтения пользователей и рекомендовать им наиболее подходящие элементы.

#### <span style="color:red"> На полный dataframe машине может не хватать ресурсов и ядро аварийно завершает работу. В связи с этим можно работать на данных ev_light. 

````python
# Создадим копию
data_features = ev.copy()
# Преобразовываем признак "event" в оценки: принимая "addtocart" или "transaction" = 1, "event" = 0
data_features["like"] = data_features.apply(lambda x: 1 if (x["event"] == "addtocart" or x["event"] == "transaction") else 0, axis=1)
data_features["like"] = data_features["like"].astype("Int32")
data_features = data_features.drop(columns=["event"], axis=1) # удаляем признак "event"
# Переставим признак "like" для удобства визуализации
cols = list(data_features.columns)
cols.insert(2, cols.pop(cols.index("like")))
data_features = data_features[cols]
````

### 4.1. Генерация факторов, связанных с товарами <a id="item-four-one"></a>

Генерация факторов, связанных с товарами, включает в себя создание признаков, которые описывают свойства элементов (товаров, услуг и т. д.) в системе. Эти факторы могут включать в себя информацию о категориях товаров, ценах, популярности, оценках пользователей и другие характеристики, которые могут быть важны для рекомендаций.

````python
# Создание разреженной матрицы user-item в формате CSR
item_user_coo = coo_matrix(
    (data_features["like"], (data_features["itemid"], data_features["visitorid"])))

item_user_matrix = csr_matrix(item_user_coo)

# Инициализация ALS модели для user-item матрицы
als_model = AlternatingLeastSquares(
    factors=15,           # регулируем количество создаваемых факторов
    regularization=0.01,  # меняем регуляризацию для исключения 0 факторов
    dtype= np.float64,    # меняем размер числа признаков
    iterations=50,        # увеличиваем количество итераций
    use_gpu=False,        # использовать GPU, если доступно
    random_state=RANDOM_SEED
)

# Обучение модели ALS для item-user матрицы
als_model.fit(item_user_matrix)

# получение факторов для item
item_factors = als_model.item_factors

# Создание DataFrame с факторами item
item_factors_df = pd.DataFrame(
    item_factors,
    index=range(1, item_factors.shape[0] + 1),
    columns=[f"item_factor_{i}" for i in range(item_factors.shape[1])]
)

# Добавление факторов предметов к исходному датасету с item 
data_features = pd.merge(
    data_features,
    item_factors_df,
    left_on="itemid",
    right_index=True,
    how="left"
)
````

##### 5.1.2.3 Подсчет, зависящий от времени / Time dependent count <a id="item-five-one-two-three"></a>

Во многих сценариях временная зависимость играет решающую роль при подготовке набора данных для построения модели совместной фильтрации, которая фиксирует изменение интересов пользователей с течением времени. Одним из распространенных методов получения счета, зависящего от времени, является добавление коэффициента затухания во времени при подсчете. Формула для получения показателя сходства для каждой пары пользователь-элемент имеет вид

$$
a_{ij}=\sum_k w_k \left(\frac{1}{2}\right)^{\frac{t_0-t_k}{T}} 
$$

Аффинность $a_{ij}$ для пользователя $i$ и товара $j$ определяется как взвешенная сумма всех $k$ событий, связанных с пользователем $i$ и товаром $j$. $w_k$ обозначает вес определенного события, а степень двойки отражает событие со скидкой во времени. Коэффициент масштабирования $(\frac{1}{2})^n$ делает параметр $T$ периодом полураспада: события, произошедшие за $T$ единиц до $t_0$, получат половину веса тех, которые произошли в момент $t_0$.

Повторяя этот расчет для всех $n$ пользователей и $m$ товаров, получается матрица размером $n \times m$, обозначаемая как $A$. Упрощение данного выражения возможно путем установки всех весов равными 1 (по сути игнорируя типы событий) либо путем задания параметра периода полураспада $T$ бесконечным (игнорирование временных меток транзакций).

Мы используем 5 дней в качестве параметра периода полураспада и используем последнее время в наборе данных в качестве временной привязки.

````python
# Создаем копию данных и проводим преобразование для full
# Добавляем столбец весов
data_weight = ev_full[["visitorid", "itemid", "event", "date"]].copy()

conditions = [
    data_weight["event"] == "view",
    data_weight["event"] == "addtocart",
    data_weight["event"] == "transaction"
]

choices = [1, 3, 5]

data_weight["weight"] = np.select(conditions, choices, default="black")

# Преобразовать в числовой тип.
data_weight["weight"] = pd.to_numeric(data_weight["weight"])

data_dependent = data_weight.copy()
# Устанавливаем велечину периода полураспада Т
T = 5
# Устанавливаем временную привязку
t_ref = pd.to_datetime(data_dependent["date"]).max()
# Проводим преобразование
data_dependent["timedecay"] = data_dependent.apply(
    lambda x: x["weight"] * np.power(0.5, (t_ref - pd.to_datetime(x["date"])).days / T), 
    axis=1
)
# Считайте с учетом веса.
data_dependent_count = data_dependent.groupby(["visitorid", "itemid"])["timedecay"].sum().reset_index()
data_dependent_count.columns = ["visitorid", "itemid", "like"]

# Выводим информацию после преобразования для оценки
data_dependent_count.head(3)
````

### 5.2. Разделение данных / Data split <a id="item-five-two"></a>

**Разделение данных** - одна из наиболее важных задач при оценке систем рекомендаций. Стратегия разделения сильно влияет на метрики оценки, поэтому всегда следует уделять ей пристальное внимание. Рассмотрим, как и где лучше всего применять различные стратегии разделения данных в контексте построения рекомендательных систем:

**1. Случайное разделение (Random Split)**

***Когда применять:***
* Когда у вас есть большой объем данных, и вы уверены, что они достаточно репрезентативны и сбалансированы.
* Когда ваши данные не имеют временной зависимости или специфических структурных особенностей.

**2. Разделение по времени (Time-Based Split)**

***Когда применять:***
* Когда ваши данные представляют собой временные ряды, такие как продажи, клики или просмотры.
* Когда вам необходимо проверить, насколько хорошо ваша модель предсказывает будущее поведение на основе прошлых данных.

**3. Стратифицированное разделение (Stratified Split)**

***Когда применять:***
* Когда классы в ваших данных распределены неравномерно (например, много положительных отзывов и мало отрицательных).
* Когда вам нужно убедиться, что тестовая выборка отражает тот же баланс классов, что и полный набор данных.

**4. Кросс-валидация (Cross-Validation)**

***Когда применять:***
* Когда у вас ограниченное количество данных, и вы хотите максимально эффективно использовать их для оценки модели.
* Когда вам нужна более надежная оценка производительности модели, основанная на множестве различных комбинаций тренировочных и тестовых данных.

**5. Leave-One-Out Cross-Validation (LOOCV)**

***Когда применять:***
* Когда у вас очень маленький набор данных, и вы хотите использовать каждый образец данных как минимум один раз в качестве тестовой выборки.
* Когда точность оценки важнее вычислительной эффективности.

### **Рекомендательные системы - кратко.**

Цель рекомендательной системы заключается в предсказании предпочтений пользователя относительно определенного набора товаров на основании прошлого опыта.
Рекомендательные системы делятся на два основных типа: "Персонализированные" и "Неперсонализированные". Разница между ними заключается в том, как они подбирают рекомендации для каждого конкретного пользователя.

**I. Персонализированные рекомендательные системы** - персонализированная система строит рекомендации на основе ***предпочтений и поведения конкретного пользователя***. Это означает, что рекомендации будут ***уникальными для каждого пользователя***, исходя из его истории взаимодействия с системой, оценок, покупок, просмотров и других действий.
* Преимущества:
    - Более точные и релевантные рекомендации, так как учитываются индивидуальные интересы и поведение пользователя.
    - Возможность адаптироваться к изменениям интересов пользователя со временем.
* Недостатки:
    - Требуется большое количество данных о пользователе для эффективной работы.
    - Может возникнуть проблема "холодного старта" для новых пользователей, у которых еще нет истории взаимодействий.

**II. Неперсонализированные рекомендательные системы** - неперсонализированные системы делают ***одинаковые рекомендации для всех пользователей*** без учета их индивидуальных предпочтений. Они обычно основываются на популярности или общих трендах среди всех пользователей.
* Преимущества:
    - Простота реализации и отсутствие необходимости собирать большие объемы данных о каждом пользователе.
    - Подходит для ситуаций, когда важно показать популярные или трендовые продукты/контент всем пользователям.
* Недостатки:
    - Рекомендации могут быть менее точными и интересными для конкретных пользователей, так как не учитывают их личные предпочтения.
    - Отсутствие возможности адаптации к интересам отдельных пользователей.

3. **Гибридные системы** - когда рекомендательные системы сочетают элементы обоих типов, создавая гибридные решения. Такие системы могут сначала предложить общие рекомендации, а затем постепенно персонализировать их на основе активности пользователя.

### **Оценка работы моделей рекомендательных систем**

Оценка качества работы модели рекомендательной системы является важным этапом её разработки и внедрения. Существует несколько ключевых метрик, которые помогают оценить эффективность таких моделей. Метрики делятся на две большие группы: точностные и ранговые. 

**Точностные метрики**

Точностные метрики оценивают точность предсказаний модели относительно истинных предпочтений пользователей. К ним относятся:

1. ***Precision (Точность)*** — доля правильных рекомендаций среди всех сделанных системой. Например, если система рекомендует 10 фильмов, а пользователь заинтересовался тремя из них, то точность будет равна 0.3.
2. ***Recall (Полнота)*** — доля найденных релевантных элементов среди всех возможных релевантных элементов. Если у пользователя есть 20 интересных ему фильмов, но модель рекомендовала всего 15, из которых он посмотрел 12, то полнота составит 12 / 20 = 0.6;
3. ***F1-score*** — гармоническое среднее между точностью и полнотой, которое позволяет сбалансированно оценивать обе эти характеристики.

**Ранговые метрики**

Ранговые метрики учитывают порядок рекомендаций, что особенно важно для рекомендательных систем, так как пользователи обычно просматривают лишь первые несколько позиций списка рекомендаций. Примеры ранговых метрик включают:
1. ***Mean Average Precision (MAP)*** — средняя точность по всем пользователям, где учитываются позиции релевантных рекомендаций;
2. ***Normalized Discounted Cumulative Gain (NDCG)*** — взвешенная сумма релевантностей рекомендуемых элементов, где элементы на первых позициях имеют больший вес.

#### Заказчиком принято решение производить оценку работы моделей с помощью метрики - ***Precision@3*** - точность по первым 3 рекомендациям.
- одной из наиболее распространенных точностных метрик является Precision@k, которая оценивает долю релевантных рекомендаций среди первых $k$ предложенных системой. Эта метрика помогает понять, насколько хорошо модель справляется с задачей предложить пользователю интересные ему объекты на начальных позициях списка рекомендаций. Формула расчета:

$$
\text{Precision}@k = \frac{|\{R_k \cap T\}|}{|R_k|} = \frac{\text{количество релевантных объектов среди первых } k}{\text{общее количество первых } k}
$$

- метрика **Precision@k** важна потому, что она отражает качество рекомендаций именно на тех позициях, которые чаще всего видит пользователь. В реальной жизни пользователи редко просматривают весь длинный список предложений, поэтому важно, чтобы самые верхние рекомендации были максимально релевантны их интересам. Это повышает вероятность того, что пользователь воспользуется предложением и останется доволен сервисом.
 
- оценка работы моделей рекомендательных систем требует комплексного подхода, включающего использование как точностных, так и ранговых метрик. Precision@k является одной из ключевых метрик, позволяющих оценить качество верхних позиций списка рекомендаций, что имеет решающее значение для успеха любой рекомендательной системы.

````python
# Проводим преобразование
X_train_start = train_xgb.drop(columns="like", axis=1)
y_train = train_xgb["like"]

X_test_start = test_xgb.drop(columns="like", axis=1)
y_test = test_xgb["like"]

# Инициируем Скаллер
scaler = StandardScaler()
scaler.fit(X_train_start)

# Нормализуем тренировочные данным
X_train = scaler.transform(X_train_start)
# Нормализуем тестовые данным
X_test = scaler.transform(X_test_start)

# Выводим данные
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
````

````python
# Оценим распределение классов в тестовой выборке
y_test.value_counts() # Распределение не равномерное, что будет влиять на качество метрики Precision_score
````

````python
# Получаем предсказания
predicts = model_xgb.predict(X_test)
# Проводим расчет метрики на тестовом наборе
precision = metrics.precision_score(y_test, predicts)
recall = metrics.recall_score(y_test, predicts)
f1 = metrics.f1_score(y_test, predicts)
accuracy = metrics.accuracy_score(y_test, predicts)
# Выводим метрики
print(f'Precision_score: {precision:.8f}.')
print(f'Recall_score: {recall:.8f}.')
print(f'f1_score: {f1:.8f}.')
print(f'Accuracy_score: {accuracy:.8f}.')
````

````python
# Преобразуем данные в DMatrix для XGBoost
dtrain = xgb.DMatrix(X_train, label=y_train)
dvalid = xgb.DMatrix(X_test, label=y_test)

# Функция для оптимизации гиперпараметров
def objective(trial):
    param = {
        "max_depth": trial.suggest_int("max_depth", 3, 100),
        "learning_rate": trial.suggest_loguniform("learning_rate", 0.01, 0.5),
        "n_estimators": trial.suggest_int("n_estimators", 100, 500),
        "gamma": trial.suggest_loguniform("gamma", 1e-8, 1.0),
        "min_child_weight": trial.suggest_loguniform("min_child_weight", 1e-8, 10.0),
        "subsample": trial.suggest_uniform("subsample", 0.6, 1.0),
        "colsample_bytree": trial.suggest_uniform("colsample_bytree", 0.6, 1.0),
        "reg_alpha": trial.suggest_loguniform("reg_alpha", 1e-8, 1.0),
        "reg_lambda": trial.suggest_loguniform("reg_lambda", 1e-8, 1.0),
        "objective": "binary:logistic",
        "eval_metric": "logloss",
    }

    # Тренировка модели
    model = xgb.train(param, dtrain, evals=[(dvalid, "validation")], verbose_eval=False)

    # Прогнозирование на проверочном наборе
    predicts = model.predict(dvalid)
    pred_labels = np.where(predicts > 0.5, 1, 0)
    accuracy = accuracy_score(y_test, pred_labels)

    return 1 - accuracy

# Запуск процесса оптимизации
study = optuna.create_study(direction="minimize")
study.optimize(objective, n_trials=10)

# Получение лучших параметров
best_params = study.best_params
print(f"Best params: {best_params}")
````

````python
# Описываем функция для создания матрицы
def create_user_item_matrix(df):
    rows = df["visitorid"].astype(np.int32)
    cols = df["itemid"].astype(np.int32)
    weights = df["like"]
    
    # Создаём разреженную матрицу
    matrix = csr_matrix((weights, (rows, cols)))
    
    return matrix

train_matrix = create_user_item_matrix(train_count)
# train_matrix = bm25_weight(train_matrix, K1=40, B=1.7)
# train_matrix = train_matrix.tocsr()
test_matrix = create_user_item_matrix(test_count)
# test_matrix = bm25_weight(test_matrix, K1=40, B=1.7)
# test_matrix = test_matrix.tocsr()
# Выводим информацию
print(f'Размер train csr_matrix: {train_matrix.shape}')
print(f'Размер test csr_matrix: {test_matrix.shape}')
````

### Что такое BM25Recommender?

Алгоритм **BM25Recommender**, реализованный в библиотеке `Implicit`, является адаптацией классического информационного поиска алгоритма **Okapi BM25** для задач рекомендательных систем. Этот алгоритм был первоначально предложен для ранжирования документов в поисковых системах, но оказался полезным и в контексте рекомендаций, особенно когда требуется учесть не только частоту взаимодействий, но и особенности самих элементов.

Алгоритм использует матрицу взаимодействий пользователей и предметов (например, покупки, клики, лайки). Эти взаимодействия обычно представлены в виде разреженной матрицы, где каждая ячейка указывает на наличие или отсутствие взаимодействия между конкретным пользователем и предметом.


***BM25 основывается на трех ключевых параметрах:***
- **IDF (Inverse Document Frequency)** – обратная частота документа. Чем реже предмет встречается в взаимодействиях, тем больший вес ему присваивается.
- **Term Frequency (TF)** – частота термина. Чем чаще предмет взаимодействует с пользователями, тем выше его значимость.
- **Field Length Normalization** – нормализация длины поля. Это помогает избежать чрезмерного влияния длинных списков взаимодействий.

***Формула BM25:***

$$ \text{BM25}(d, q) = \sum_{t \in q} \text{IDF}(t) \cdot \frac{\text{TF}(t, d) \cdot (k_1 + 1)}{\text{TF}(t, d) + k_1 \cdot (1 - b + b \cdot \frac{|d|}{\text{avgdl}})}

$$
где:
- $ t $ – термин (предмет);
- $ d $ – документ (профиль пользователя);
- $ q $ – запрос (интересующий нас профиль пользователя);
- $ |d| $ – длина документа (количество взаимодействий пользователя);
- $ \text{avgdl} $ – средняя длина документа;
- $ k_1 $ и $ b $ – параметры настройки, которые контролируют влияние частоты терминов и нормализации длины соответственно.

***В контексте рекомендаций:***
- Пользователи рассматриваются как документы.
- Предметы (товары, фильмы и т.п.) являются терминами.
- Цель состоит в том, чтобы найти предметы, которые наиболее релевантны профилю конкретного пользователя.

***Особенности и преимущества BM25Recommender:***
- **Учитывает редкость предметов**: Алгоритм придает больший вес редко встречающимся предметам, что полезно в ситуациях, когда редкие товары могут быть более интересными для конкретных пользователей.
- **Нормализует длину профиля пользователя**: Избегает смещения в сторону пользователей с большим количеством взаимодействий.
- **Простота реализации и интерпретации**: Алгоритм относительно прост в понимании и применении, что делает его удобным для использования в реальных приложениях.

Таким образом, **BM25Recommender** представляет собой эффективный инструмент для создания персонализированных рекомендаций, основанный на адаптации известного алгоритма информационного поиска для задач рекомендательных систем.

**BM25Recommender** сам по себе не является классическим алгоритмом KNN, его применение в контексте рекомендательных систем делает его частью семейства моделей ближайшего соседа, поскольку он используется для нахождения близких (по смыслу) предметов на основе взаимодействий пользователей.

**Почему BM25 считается моделью ближайшего соседа?**
1. Сходство между предметами: В контексте рекомендательных систем BM25 измеряет степень похожести между предметами на основе взаимодействий пользователей. То есть, если два предмета часто встречаются вместе в профилях пользователей, они считаются близкими (или похожими) по смыслу.
2. Использование расстояний/сходств: Модели ближайшего соседа работают с понятием расстояния или сходства между объектами. В случае BM25, хотя формула основана на частотах взаимодействий и нормализации, конечный результат можно интерпретировать как меру близости между предметами.
3. Рекомендация на основе сходства: Когда BM25 используется для рекомендаций, он находит предметы, которые наиболее близки к уже известным предпочтениям пользователя, что соответствует принципу работы моделей ближайшего соседа.

<span style="color:red">Необходимо обратить внимание, что в библиотеке отсутсвует встроенная функция метрики Precision_at_k, но представлен пример кода для расчета и использования. При этом в функции есть настраиваемый параметр treshold, который необходимо настраивать в зависимости от данных. При работе на разных данных были выставлены параметры: weight / count = 3.5 (как указано в документации), expert - 0.5, и это может затруднить сравнение результатов работы алгоритмов с другими библиотеками.

````python
# Функция для расчета метрики precision@k и recall@k из библиотеки Surprise
def precision_recall_at_k(predictions, k=3, threshold=3.5):
    """Функция вычисления метрики precision@k и recall@k
    
    Args:
        predictions (dict): словарь с результатами предсказаний модели.
        k (int, optional): количество рекомендаций / товар по которым необходимо произвести оценку. По умолчанию = 3.
        threshold (float, optional): минимальный уровень достоверности, который должен быть достигнут, 
                                     чтобы элемент считался значимым для расчета метрик точности и полноты. По умолчанию = 0.5

    Returns:
        precisions (float): расчитаная метрика precision@k на выборке
        recalls (float): расчитаная метрика recall@k на выборке
    """
    from collections import defaultdict 

    # First map the predictions to each user.
    user_est_true = defaultdict(list)
    for uid, _, true_r, est, _ in predictions:
        user_est_true[uid].append((est, true_r))

    precisions = dict()
    recalls = dict()
    for uid, user_ratings in user_est_true.items():

        # Sort user ratings by estimated value
        user_ratings.sort(key=lambda x: x[0], reverse=True)

        # Number of relevant items
        n_rel = sum((true_r >= threshold) for (_, true_r) in user_ratings)

        # Number of recommended items in top k
        n_rec_k = sum((est >= threshold) for (est, _) in user_ratings[:k])

        # Number of relevant and recommended items in top k
        n_rel_and_rec_k = sum(
            ((true_r >= threshold) and (est >= threshold))
            for (est, true_r) in user_ratings[:k]
        )

        # Precision@K: Proportion of recommended items that are relevant
        # When n_rec_k is 0, Precision is undefined. We here set it to 0.

        precisions[uid] = n_rel_and_rec_k / n_rec_k if n_rec_k != 0 else 0

        # Recall@K: Proportion of relevant items that are recommended
        # When n_rel is 0, Recall is undefined. We here set it to 0.

        recalls[uid] = n_rel_and_rec_k / n_rel if n_rel != 0 else 0
        
        # усредняем Precision и recall по всем пользователям
        precisions_mean = sum(prec for prec in precisions.values()) / len(precisions)
        recalls_mean = sum(rec for rec in recalls.values()) / len(recalls)

    return precisions_mean, recalls_mean
````

````python
# Функция для вывода рекомендаций для клиента - преобразованная для формирования df
def get_top_n_modified(predictions, n=3, user_uid=None):
    """Возвращает топ-n рекомендации для каждого пользователя из списка прогнозов.

    Аргументы:
        model (str): Наименование модели для формирования предсказаний.
        data (surprise.dataset): Набор данных, используемый для формирования предсказаний.
        n (int): Количество рекомендаций, которые нужно вернуть для каждого пользователя. По умолчанию 3.
        user_uid (optional int): Номер пользователя, для которого нужно вывести рекомендации. По умолчанию None.

    Возврат:
        Словарь, в котором ключи — уникальные Id пользователей, а значения — списки пар: item_id, est(расчетный рейтинг), содержащие топ-n рекомендации для каждого пользователя.
    """
    from collections import defaultdict
    
    top_n = defaultdict(list)
    
    #predictions = model.test(testset)
    
    for uid, iid, true_r, est, details in predictions:
        if user_uid is None or uid == user_uid:
            top_n[uid].append((iid, est)) # можно добавлять true_r, est, details

    # Формирование рекомендаций
    for uid, user_ratings in top_n.items():
        user_ratings = list(user_ratings)  # Преобразование set в список
        user_ratings.sort(key=lambda x: x[1], reverse=True)
        top_n[uid] = user_ratings[:n]

    item_recomendations = []
    # Если у пользователя нет истории покупок, выводим предупреждение
    if len(top_n[user_uid]) == 0:
        return print(f'Пользователь c UserId = {user_uid} не имеет истории покупок, мы ему можем рекомендовать ТОП-{n} покупаемых товаров!')
    else:
        sorted_recommendations = sorted(top_n[user_uid], key=lambda x: x[1], reverse=True)
        # print(f'Пользователю с UserId = {user_uid} рекомендованы товары:')
        for item_id, rating in sorted_recommendations:
            #print(f'Товар с ItemId = {item_id}, рейтинг = {rating:.2f}')
            item_recomendations.append(item_id)
    
    return item_recomendations
````

### Что такое SVDpp?

Алгоритм **SVD++** (Singular Value Decomposition++) является расширением базовой модели **SVD** (Singular Value Decomposition) и был впервые представлен в статье «Factorization Meets the Neighborhood: a Multifaceted Collaborative Filtering Model» авторов Yehuda Koren и Robert Bell. Этот алгоритм стал популярным благодаря своей эффективности в задачах коллаборативной фильтрации и широко используется в рекомендательных системах.

**Основная идея SVD++** — улучшить предсказательную способность стандартной модели SVD путем учета информации о взаимодействии пользователей с элементами, которые они уже оценили. Стандартная модель SVD рассматривает только скрытые факторы пользователей и элементов, тогда как SVD++ добавляет информацию о взаимодействии пользователя с другими элементами через механизм имплицитной обратной связи.

**Математическая формулировка** - модель SVD++ описывает предсказание рейтинга $ r_{ui} $ для пользователя $ u $ и элемента $ i $ следующим образом:

$$
\hat{r}_{ui} = \mu + b_u + b_i + q_i^T(p_u + |I_u|^{−\frac{1}{2}}\sum_{j\in I_u}y_j)

$$

Где:
- $\mu$ — глобальное среднее значение всех рейтингов,
- $b_u$ — смещение пользователя $u$,
- $b_i$ — смещение элемента $i$,
- $q_i$ — вектор скрытого фактора элемента $i$,
- $p_u$ — вектор скрытого фактора пользователя $u$,
- $|I_u|$ — количество элементов, с которыми взаимодействовал пользователь $u$,
- $y_j$ — вектор скрытого фактора элемента $j$, с которым взаимодействовал пользователь $u$.


Основное отличие **SVD++** от стандартного **SVD** заключается в добавлении члена $|I_u|^{−\frac{1}{2}}\sum_{j\in I_u}y_j$. Этот член учитывает взаимодействие пользователя с другими элементами, что позволяет лучше моделировать предпочтения пользователя на основе его истории взаимодействий.

### 8.1. Типы рекомендательных систем (РС)

#### **<span style="color:red">I. Персонализированные рекомендательные системы (РС)** - делятся на несколько категорий в зависимости от подхода, который используется для генерации рекомендаций. Основные типы персонализированных РС включают:

##### 1. **Контентная фильтрация (Content-Based Filtering)**

Этот подход опирается на характеристики самого предмета (фильма, товара, статьи и т.п.) и предпочтения пользователя. Контентная фильтрация предполагает следующие шаги:
- Анализируются характеристики предметов (жанр фильма, автор книги, ингредиенты блюда и т.д.).
- Оцениваются предпочтения пользователя на основе его прошлых взаимодействий с предметами.
- Предметы, наиболее соответствующие этим предпочтениям, рекомендуются пользователю.

##### 2. **Коллаборативная фильтрация (Collaborative Filtering)**

Этот метод основан на предположении, что пользователи с похожими вкусами будут иметь схожее мнение о продуктах или контенте. Коллаборативная фильтрация делится на две основные подгруппы:

- **Фильтрация на основе пользователей (User-based CF)**:
  - Система ищет пользователей с похожим поведением (например, кто смотрел те же фильмы или купил те же товары) и рекомендует им те вещи, которые понравились другим пользователям с аналогичными вкусами.
  
- **Фильтрация на основе предметов (Item-based CF)**:
  - Система находит предметы, которые похожи друг на друга (например, книги одного автора или фильмы одного жанра), и рекомендует их пользователю на основании тех вещей, которые он уже оценил положительно.

##### 3. **Матричное разложение (Matrix Factorization)**

Это статистический метод, который применяется для прогнозирования неизвестных значений в большой матрице данных. В контексте рекомендательных систем матрица содержит рейтинги пользователей по предметам. Матрица разлагается на две меньшие матрицы, которые представляют собой латентные факторы пользователей и предметов. Этот метод часто используется в сочетании с другими алгоритмами для повышения точности рекомендаций.

##### 4. **Гибридные системы (Hybrid Systems)**

Гибридные системы объединяют несколько подходов для достижения наилучшего результата. Например, можно комбинировать коллаборативную фильтрацию с контентной фильтрацией, чтобы учесть как вкусы пользователя, так и характеристики самих предметов. Гибридные системы позволяют компенсировать недостатки одних методов за счет преимуществ других.

##### 5. **Методы машинного обучения (Machine Learning Approaches)**

Современные рекомендательные системы все чаще используют методы машинного обучения, такие как:
- **Кластеризация**: Группировка пользователей или предметов на основе их характеристик. Пользователи внутри одного кластера получают схожие рекомендации.
- **Классификация и регрессия**: Использование моделей для предсказания вероятности того, что пользователь заинтересуется определенным предметом.
- **Нейронные сети**: Глубокое обучение и использование нейронных сетей для выявления сложных взаимосвязей между пользователями и предметами.

##### 6. **Модель Knowledge Graph (Граф знаний)**

Этот подход основан на использовании графовых структур для представления отношений между объектами (пользователями, предметами, характеристиками и т.д.). Граф знаний помогает учитывать сложные связи между элементами и генерировать более точные рекомендации.

#### **<span style="color:red">II Неперсонализированные рекомендательные системы (РС)** - делятся на несколько категорий в зависимости от стратегии подбора рекомендаций. Вот основные типы таких систем:

##### 1. **Популярные рекомендации (Popularity-Based Recommendations)**

Этот подход основан на предоставлении пользователям самых популярных товаров или контента. Популярность определяется количеством продаж, просмотров, лайков или других показателей вовлеченности. Например:
- Топ-продаваемые товары в магазине.
- Самые просматриваемые видео на платформе видеохостинга.
- Наиболее читаемые статьи на новостном сайте.

##### 2. **Новинки (New Items Recommendations)**

В этом случае рекомендуется недавно добавленный контент или товары. Это полезно для привлечения внимания к новым продуктам или обновлениям. Например:
- Новые поступления в книжный магазин.
- Свежие выпуски сериалов на стриминговой платформе.
- Новинки сезона в модном бутике.

##### 3. **Трендовые рекомендации (Trending Recommendations)**

Здесь предлагаются товары или контент, которые находятся на пике популярности в данный момент времени. Тренды могут меняться быстро, поэтому этот подход актуален для динамичных платформ. Например:
- Тенденции в социальных сетях.
- Горячие темы обсуждения в новостных изданиях.
- Хиты музыкальных чартов.

##### 4. **Категоризированные рекомендации (Category-Based Recommendations)**

Этот метод предлагает пользователям контент или товары, относящиеся к определённой категории. Пользователь выбирает категорию, и система показывает ему предложения из этой категории. Например:
- Книги по жанрам (детективы, фантастика, романы).
- Фильмы по категориям (комедии, драмы, ужасы).
- Товары по типу (электроника, одежда, косметика).

##### 5. **Местоположение (Location-Based Recommendations)**

Этот подход учитывает географическое положение пользователя и предлагает ему местные события, заведения или услуги. Например:
- Рестораны поблизости.
- Мероприятия в городе.
- Магазины в радиусе нескольких километров.

##### 6. **Время суток/день недели (Time-Based Recommendations)**

Система может предлагать разные рекомендации в зависимости от времени суток или дня недели. Например:
- Утренние новости.
- Вечерний кинопоказ.
- Обеды выходного дня.

##### 7. **Демографические рекомендации (Demographic-Based Recommendations)**

Этот метод делает рекомендации на основе демографических характеристик пользователей, таких как возраст, пол, образование и т.д. Например:
- Детская литература для младшей возрастной группы.
- Продукты для пожилых людей.
- Курсы профессионального развития для студентов.

##### 8. **Рекомендации на основе сезонности (Seasonal Recommendations)**

Такие рекомендации зависят от времени года или праздников. Например:
- Летняя коллекция одежды.
- Рождественские подарки.
- Зимнее спортивное снаряжение.

##### 9. **Общие рекомендации (General Recommendations)**

Это универсальные рекомендации, которые подходят большинству пользователей и не зависят от каких-либо специфичных факторов. Например:
- Классические произведения литературы.
- Бестселлеры электроники.
- Основополагающие курсы по программированию.

### <span style="color:blue">**2. Библиотека Implicit**.


**Implicit** — это библиотека Python, которая предназначена для работы с неявными (implicit) предпочтениями пользователей в рекомендательных системах. Она была разработана Бенджамином Картером (Benjamin Carter) и активно поддерживается сообществом. Основное назначение этой библиотеки заключается в предоставлении инструментов для работы с методами факторизационных машин (factorization machines) и других алгоритмов, используемых для обработки неявных предпочтений.

Библиотека implicit предназначена для реализации рекомендательных систем и основана на факторизации матриц. Она особенно полезна для задач коллаборативной фильтрации, таких как рекомендации товаров пользователям на основе их предыдущих взаимодействий с этими товарами.

Основные особенности библиотеки Implicit:
1. ***Неявные предпочтения:***
    * В отличие от явных предпочтений (explicit preferences), когда пользователи прямо выражают свою оценку объектов (например, ставят звездочки фильмам), неявные предпочтения подразумеваются косвенно через поведение пользователя (например, клики, просмотры страниц, добавление в корзину).     * Alternating Least Squares (ALS): Алгоритм чередующихся наименьших квадратов, Bayesian Personalized Ranking (BPR): Байесовский персонализированный ранжир, Logistic Matrix Factorization (LMF): Логистическая факторизация матриц.
3. ***Оптимизированная производительность:***
    * Implicit написана на Python, но критические участки кода реализованы на Cython, что значительно ускоряет выполнение вычислений. Это делает библиотеку подходящей для работы с большими наборами данных;
    * Поддержка многопоточных вычислений.
4. ***Простота использования:***
    * Как и многие современные библиотеки для машинного обучения, Implicit предоставляет удобный интерфейс для подготовки данных, обучения моделей и получения рекомендаций. Это упрощает интеграцию библиотеки в существующие проекты.

### 8.3. Дополнительная информация по алгоритмам / методам

**Меры сходимости**

В коде `'msd'` является названием меры сходства (similarity measure), которая используется в моделях коллаборативной фильтрации, таких как `KNNBasic`, `KNNWithMeans`, `KNNWithZScore` и `KNNBaseline`. В библиотеке Surprise доступны различные меры сходства, и каждая из них может влиять на качество рекомендаций.

Аббревиатура **MSD** расшифровывается как **Mean Squared Difference** (Среднеквадратичное различие). Эта мера сходства измеряет степень различия между двумя пользователями или элементами на основе их рейтингов. Формула MSD выглядит следующим образом:

$$
\text{MSD}(u, v) = \frac{\sum_{i \in I_{uv}} (r_{ui} - r_{vi})^2}{|I_{uv}|}

$$

где:
- $ u $ и $ v $ — пользователи или элементы,
- $ I_{uv} $ — множество элементов, которые были оценены обоими пользователями,
- $ r_{ui} $ и $ r_{vi} $ — рейтинги пользователей $ u $ и $ v $ соответственно для элемента $ i $,
- $ |I_{uv}| $ — количество общих оценённых элементов.

Чем меньше значение MSD, тем более схожи пользователи или элементы. Таким образом, использование MSD в качестве меры сходства помогает находить наиболее похожих пользователей или элементы для формирования рекомендаций.

Другие возможные меры сходства включают Pearson correlation, cosine similarity и другие. Выбор конкретной меры зависит от специфики задачи и особенностей ваших данных.

Помимо **Mean Squared Difference (MSD)**, библиотека Surprise предлагает несколько других мер сходства, которые могут использоваться в моделях коллаборативной фильтрации. Вот некоторые из них:

1. **Pearson Correlation (коэффициент корреляции Пирсона)**:

$$
   \text{Pearson}(u, v) = \frac{\sum_{i \in I_{uv}} (r_{ui} - \bar{r}_u)(r_{vi} - \bar{r}_v)}{\sqrt{\sum_{i \in I_{uv}} (r_{ui} - \bar{r}_u)^2} \sqrt{\sum_{i \in I_{uv}} (r_{vi} - \bar{r}_v)^2}}

$$
   где:
   - $ \bar{r}_u $ и $ \bar{r}_v $ — средние рейтинги пользователей $ u $ и $ v $ соответственно.

   Коэффициент Пирсона оценивает линейную зависимость между рейтингами двух пользователей или элементов. Значения варьируются от $-1$ до $+1$, где $+1$ обозначает идеальную положительную корреляцию, а $-1$ — идеальную отрицательную корреляцию.

2. **Cosine Similarity (косинусная мера сходства)**:

$$
   \text{Cosine}(u, v) = \frac{\sum_{i \in I_{uv}} r_{ui}r_{vi}}{\sqrt{\sum_{i \in I_{uv}} r_{ui}^2} \sqrt{\sum_{i \in I_{uv}} r_{vi}^2}}

$$

   Косинусная мера сходства определяет угол между векторами рейтингов двух пользователей или элементов. Значение варьируется от $0$ до $1$, где $1$ соответствует полной схожести, а $0$ — отсутствию схожести.

3. **Jaccard Index (индекс Жаккара)**:

$$
   \text{Jaccard}(u, v) = \frac{|I_u \cap I_v|}{|I_u \cup I_v|}

$$

   Индекс Жаккара сравнивает множества объектов, оценённых двумя пользователями. Он равен отношению количества общих оценённых объектов к общему количеству уникальных объектов, оценённых обоими пользователями. Значение находится в диапазоне от $0$ до $1$.

4. **Adjusted Cosine Similarity (скорректированная косинусная мера сходства)**:

$$
   \text{AdjCosine}(u, v) = \frac{\sum_{i \in I_{uv}} (r_{ui} - \mu_i)(r_{vi} - \mu_i)}{\sqrt{\sum_{i \in I_{uv}} (r_{ui} - \mu_i)^2} \sqrt{\sum_{i \in I_{uv}} (r_{vi} - \mu_i)^2}}

$$

   Скорректированная косинусная мера учитывает разницу между средними значениями рейтингов элементов, что делает её менее чувствительной к смещению средних значений рейтингов разных пользователей.

Эти меры могут применяться в зависимости от типа данных и требований вашей задачи. Например, если ваша цель — найти схожесть между пользователями на основе их предпочтений, то может подойти коэффициент корреляции Пирсона или косинусная мера сходства.

**Модель SVDpp из библиотеки Suprice**

Модель **SVD++** является одной из наиболее популярных моделей для рекомендаций на основе коллаборативной фильтрации. Она была предложена в статье *Factorization Meets the Neighborhood: a Multifaceted Collaborative Filtering Model* Йехудой Кореном в 2008 году. Эта модель представляет собой расширение базовой модели **Singular Value Decomposition (SVD)** за счет добавления информации о пользовательских оценках, основанных на взаимодействиях с другими пользователями. Давайте разберем её подробнее.

### Основная идея

Основная цель любой рекомендательной системы заключается в предсказании оценки, которую пользователь может поставить тому или иному товару/фильмам/контенту. В случае SVD++ используется информация об оценках пользователей, а также данные о том, какие товары были оценены этими пользователями ранее.

### Формула SVD++

Формула для прогнозирования оценки $ \hat{r}_{ui} $ (оценка пользователя $ u $ для товара $ i $) выглядит следующим образом:

$$
\hat{r}_{ui} = \mu + b_u + b_i + q_i^T(p_u + |I_u|^{-0.5}\sum_{j \in I_u}y_j)

$$

Где:
- $\mu$ – глобальное среднее значение всех оценок,
- $b_u$ – смещение для пользователя $u$,
- $b_i$ – смещение для товара $i$,
- $q_i$ – вектор скрытого фактора для товара $i$,
- $p_u$ – вектор скрытого фактора для пользователя $u$,
- $|I_u|$ – количество товаров, оцененных пользователем $u$,
- $y_j$ – вектор скрытого фактора для товара $j$, который был оценен пользователем $u$.

Таким образом, формула учитывает как базовые параметры, так и влияние предыдущих оценок пользователя на текущие прогнозы.

### Что нового добавляет SVD++?

В отличие от стандартной модели SVD, которая рассматривает только скрытые факторы пользователей и товаров, SVD++ дополнительно включает информацию о тех товарах, которые уже были оценены данным пользователем. Это позволяет лучше учитывать вкусы и предпочтения каждого конкретного пользователя.

### Преимущества SVD++

1. **Учет истории оценок**: Модель использует историю оценок пользователя, чтобы улучшить точность прогнозов.
2. **Универсальность**: Подходит для широкого спектра задач, связанных с рекомендациями.
3. **Высокая точность**: Благодаря учету дополнительных факторов, SVD++ часто показывает лучшие результаты по сравнению с базовыми моделями.

### Недостатки SVD++

1. **Требовательность к вычислительным ресурсам**: Из-за необходимости обработки большого количества данных, особенно при большом количестве пользователей и товаров, обучение модели может занимать значительное время.
2. **Необходимость регуляризации**: Для предотвращения переобучения требуется использовать методы регуляризации, такие как L2-регуляризация.

### Применение в Surprise

Библиотека **Surprise**, разработанная для работы с рекомендательными системами, поддерживает реализацию модели SVD++, что делает её удобным инструментом для тестирования и использования этой модели в реальных приложениях.

**Матричная факторизация**

**Матричная факторизация**- это процесс разложения матрицы на произведение более простых матриц, каждая из которых представляет собой отдельный аспект или фактор исходной матрицы. Этот процесс аналогичен факторизации чисел и полиномов, когда исходный объект разбивается на более простые составляющие. Целью матричной факторизации является упрощение сложных структур данных и выявление скрытых закономерностей. Существуют разные виды матричных разложений, каждый из которых имеет свои особенности и ограничения. Например, ранговая факторизация используется для вычисления псевдообратной матрицы и решения систем линейных уравнений.

Перечисленные алгоритмы — **SVD (Singular Value Decomposition)**, **ALS (Alternating Least Squares)**, **NMF (Non-negative Matrix Factorization)** и многие другие — действительно относятся к различным видам матричной факторизации. Матричная факторизация — это общий термин, обозначающий процесс разложения матрицы на произведение более простых матриц, каждая из которых отражает определенные аспекты исходной матрицы.

Давайте кратко рассмотрим, как каждый из перечисленных вами алгоритмов вписывается в концепцию матричной факторизации:

1. **SVD (Singular Value Decomposition)**:
   - Разложение матрицы $A$ на три матрицы: $A = U\Sigma V^T$, где $U$ и $V$ — ортонормальные матрицы, а $\Sigma$ — диагональная матрица сингулярных значений. Это классическая форма матричной факторизации, которая находит широкое применение в науке о данных и анализе сигналов.

2. **ALS (Alternating Least Squares)**:
   - Используется для решения задачи разложения матрицы на два фактора: $A \approx X Y^T$, где $X$ и $Y$ — матрицы факторов. Процесс состоит из чередующихся шагов минимизации ошибок по каждому из факторов, отсюда и название "альтернативные наименьшие квадраты". ALS активно применяется в рекомендательных системах.

3. **NMF (Non-negative Matrix Factorization)**:
   - Разложение матрицы $A$ на две неотрицательные матрицы: $A \approx W H$, где $W$ и $H$ — неотрицательные матрицы. NMF накладывает условие неотрицательности на элементы матриц, что полезно в случаях, когда нужно сохранить физическую осмысленность компонентов, например, в обработке изображений или текстов.

4. **PCA (Principal Component Analysis)**:
   - Можно рассматривать как частный случай SVD, где вместо всей матрицы сингулярных значений берётся лишь часть, соответствующая наибольшим собственным значениям. PCA находит главные компоненты, которые объясняют большую часть дисперсии данных.

5. **LDA (Linear Discriminant Analysis)**:
   - Также связано с матричным разложением, но используется для классификации, находя направления, максимизирующие разделение классов.

6. **CUR-декомпозиция**:
   - Разложение матрицы $A$ на произведение трёх матриц: $C$, $U$ и $R$, где $C$ — подматрица столбцов $A$, $R$ — подматрица строк $A$, а $U$ — промежуточная матрица. CUR-декомпозиция полезна для сжатия данных и сохранения основных свойств исходной матрицы.

Все эти методы преследуют общую цель — упростить анализ данных путём выделения важных факторов или компонентов, лежащих в основе наблюдаемых данных.

Помимо уже упомянутых **SVD**, **ALS**, **NMF** и **PCA**, существует множество других алгоритмов, относящихся к матричной факторизации. Вот некоторые из них:

1. **CP-декомпозиция (Canonical Polyadic Decomposition)**
   - CP-декомпозиция (также известная как PARAFAC) используется для разложения многомерных массивов (тензоров) на сумму внешних произведений факторов. Это особенно полезно в задачах анализа многомерных данных, таких как временные ряды, сенсорные данные и т.п.

2. **Тензорная декомпозиция Таккера (Tucker Decomposition)**
   - Подобно CP-декомпозиции, Tucker-декомпозиция также предназначена для разложения многомерных данных, но она допускает наличие корреляций между факторами, что делает её более гибкой по сравнению с CP.

3. **ICA (Independent Component Analysis)**
   - ICA используется для разделения смеси сигналов на независимые компоненты. Она находит такие компоненты, которые статистически независимы друг от друга. ICA часто применяется в обработке аудиосигналов и изображений.

4. **LDA (Latent Dirichlet Allocation)**
   - LDA — это вероятностная модель тематической кластеризации текста. Она предполагает, что каждое слово в документе генерируется из скрытой темы, а каждая тема представлена распределением вероятности над словами. LDA можно рассматривать как разновидность матричной факторизации, где документы разлагаются на темы и слова.

5. **Factor Analysis (Факторный анализ)**
   - Факторный анализ — это статистический метод, который пытается объяснить вариацию в наблюдаемых переменных через меньшее число ненаблюдаемых факторов. Этот метод также может рассматриваться как вид матричной факторизации.

6. **Matrix Completion (Завершение матрицы)**
   - Завершение матрицы — это техника, использующаяся для восстановления пропущенных значений в частично заполненной матрице. Один из подходов к завершению матрицы основывается на матричном разложении, таком как SVD или NMF.

7. **QR-декомпозиция**
   - QR-декомпозиция — это разложение матрицы $A$ на произведение ортогональной матрицы $Q$ и верхней треугольной матрицы $R$. Этот метод часто используется для решения систем линейных уравнений и нахождения собственных значений.

8. **LU-декомпозиция**
   - LU-декомпозиция — это разложение матрицы $A$ на произведение нижней треугольной матрицы $L$ и верхней треугольной матрицы $U$. Этот метод также используется для решения систем линейных уравнений и нахождения определителя матрицы.

9. **Cholesky-декомпозиция**
   - Cholesky-декомпозиция применима к симметричным положительно определенным матрицам и разлагает их на произведение нижней треугольной матрицы $L$ и её транспонированной $L^T$. Этот метод эффективен для решения систем линейных уравнений с такими матрицами.

10. **Schur-декомпозиция**
   - Schur-декомпозиция — это разложение квадратной матрицы $A$ на произведение унитарной матрицы $Q$ и верхней треугольной матрицы $T$, такой что $A = QTQ^*$. Этот метод полезен для нахождения собственных значений и векторов матрицы.

Каждый из этих методов предназначен для разных целей и условий, и выбор подходящего метода зависит от конкретной задачи и особенностей данных.

**Ранговые метрики**

#### Полный список ранговых метрик, которые подходят для оценки рекомендательных систем:

1. **Mean Average Precision (MAP)**  
   - **Описание:** Средняя точность по всем пользователям, где учитываются позиции релевантных рекомендаций.
   - **Формула:**

$$ \text{MAP} = \frac{1}{N} \sum_{i=1}^{N} AP_i

$$
   где $N$ — количество пользователей, а $AP_i$ — средняя точность для $i$-го пользователя.

2. **Normalized Discounted Cumulative Gain (NDCG)**  
   - **Описание:** Взвешенная сумма релевантностей рекомендуемых элементов, где элементы на первых позициях имеют больший вес.
   - **Формула:**

$$ \text{NDCG}_p = \frac{DCG_p}{IDCG_p}

$$
   где $DCG_p$ — Discounted Cumulative Gain для первых $p$ позиций, а $IDCG_p$ — идеальный DCG для тех же позиций.

3. **Mean Reciprocal Rank (MRR)**  
   - **Описание:** Средний обратный ранг первого релевантного элемента в списке рекомендаций.
   - **Формула:**

$$ \text{MRR} = \frac{1}{N} \sum_{i=1}^{N} \frac{1}{\text{Rank}_i}

$$
   где $N$ — общее количество запросов, а $\text{Rank}_i$ — ранг первого релевантного элемента для $i$-го запроса.

4. **Discounted Cumulative Gain (DCG)**  
   - **Описание:** Аналогичен NDCG, но без нормализации. Учитывает позицию релевантных элементов в списке рекомендаций.
   - **Формула:**

$$ \text{DCG}_p = \sum_{i=1}^p \frac{2^{rel_i} - 1}{\log_2(i+1)}

$$
   где $rel_i$ — релевантность $i$-го элемента, а $p$ — количество позиций, которое рассматривается.

5. **Rank-Biased Overlap (RBO)**  
   - **Описание:** Оценивает степень совпадения двух упорядоченных списков, принимая во внимание важность порядка элементов.
   - **Формула:**

$$ \text{RBO}(\alpha, p) = (1-\alpha) \sum_{d=1}^{\infty} \alpha^{d-1} A_d

$$
   где $\alpha$ — параметр, контролирующий значимость глубины сравнения, а $A_d$ — доля совпадающих элементов на глубине $d$.

6. **Expected Reciprocal Rank (ERR)**  
   - **Описание:** Ожидаемый обратный ранг, который учитывает вероятность того, что пользователь выберет конкретный элемент из списка рекомендаций.
   - **Формула:**

$$ \text{ERR} = \sum_{i=1}^{k} \frac{1}{i} P_i \prod_{j=1}^{i-1} (1 - P_j)

$$
   где $P_i$ — вероятность того, что пользователь выберет $i$-й элемент.

7. **Half-Life Utility Metric (HLUM)**  
   - **Описание:** Мера полезности списка рекомендаций, которая учитывает уменьшение интереса пользователя к элементам по мере продвижения вниз по списку.
   - **Формула:**

$$ \text{HLUM} = \sum_{i=1}^{k} r_i \left(\frac{1}{2}\right)^{\frac{i-1}{h}}

$$
   где $r_i$ — полезность $i$-го элемента, а $h$ — параметр, определяющий скорость убывания интереса.

Эти метрики специально разработаны для оценки качества ранжирования в рекомендательных системах, уделяя особое внимание порядку, в котором рекомендации представлены пользователю.

**Векторизация свойст товаров и создание нового DataFrame.**

```
tfidfvec = TfidfVectorizer(min_df=500, max_df=0.7)
vectorized_data = tfidfvec.fit_transform(items_pr["value"])

item_property_df = pd.DataFrame(vectorized_data.toarray(),
                       columns=tfidfvec.get_feature_names_out())
item_property_df.index = items_pr.index
display(item_property_df) 
```

1. **Создание экземпляра TfidfVectorizer**:
   ```python
   tfidfvec = TfidfVectorizer(min_df=500, max_df=0.7)
   ```
   Здесь создается объект класса `TfidfVectorizer` с параметрами:
   - `min_df=500`: минимальный порог частоты появления терма в документах. Терм будет включен в модель, если он встречается хотя бы в 500 документах.
   - `max_df=0.7`: максимальный порог частоты появления терма в документах. Терм будет исключен из модели, если он встречается более чем в 70% всех документов.

2. **Векторизация данных**:
   ```python
   vectorized_data = tfidfvec.fit_transform(items_pr["value"])
   ```
   Метод `.fit_transform()` обучает модель на данных из столбца `"value"` в DataFrame `items_pr` и сразу же применяет её для получения векторизованных данных. Результат представляет собой матрицу TF-IDF значений.

3. **Создание нового DataFrame с результатами**:
   ```python
   item_property_df = pd.DataFrame(vectorized_data.toarray(), 
                                    columns=tfidfvec.get_feature_names_out())
   ```
   - `vectorized_data.toarray()`: Преобразование sparse-матрицы в обычный массив NumPy.
   - `columns=tfidfvec.get_feature_names_out()`: Названия колонок берутся из уникальных термов, найденных в тексте. Эти термины являются ключевыми словами, используемыми для создания TF-IDF-векторов.

4. **Присвоение индекса**:
   ```python
   item_property_df.index = items_pr.index
   ```
   Индексы нового DataFrame устанавливаются равными индексам исходного DataFrame `items_pr`.

5. **Отображение результата**:
   ```python
   display(item_property_df)
   ```
   Отображается созданный DataFrame.

Итак, весь этот код выполняет следующие задачи:
- Создает модель TF-IDF с определенными параметрами.
- Обучает эту модель на данных из столбца `"value"` DataFrame `items_pr`.
- Получает векторизованные данные и создает новый DataFrame, содержащий эти данные.
- Присваивает индексы новому DataFrame и отображает его содержимое.

Параметры `min_df` и `max_df` играют важную роль в процессе векторизации текста с использованием TF-IDF. Они помогают фильтровать редко встречающиеся и слишком частые термины соответственно, улучшая качество модели. Рекомендованные значения могут варьироваться в зависимости от специфики задачи и набора данных, но вот общие рекомендации:

### min_df (Минимальная частота документа):
- **Значение по умолчанию:** 1 (терм должен встречаться хотя бы в одном документе).
- **Рекомендованное значение:** Обычно рекомендуется устанавливать значение от 2 до 10, в зависимости от размера корпуса текстов. Если корпус небольшой, то можно использовать меньшие значения (например, 2–5), чтобы избежать исключения полезных термов. Для больших корпусов можно увеличить значение до 20–50, чтобы исключить редкие термины, которые вряд ли несут значимую информацию.

### max_df (Максимальная частота документа):
- **Значение по умолчанию:** 1.0 (термы, встречающиеся во всех документах, будут включены).
- **Рекомендованное значение:** Часто используется диапазон от 0.6 до 0.9. Значения ближе к 0.6 позволяют включать термины, которые встречаются примерно в 60% документов, тогда как значения около 0.9 допускают включение даже очень частых термов. Если нужно убрать самые распространенные стоп-слова, можно установить значение ниже 0.8.

- Для небольших наборов данных: `min_df=2`, `max_df=0.8`
- Для средних наборов данных: `min_df=5`, `max_df=0.75`
- Для крупных наборов данных: `min_df=20`, `max_df=0.7`

Эти параметры всегда следует настраивать экспериментально, проверяя результаты на тестовых данных, поскольку оптимальные значения зависят от конкретной задачи и характеристик вашего набора данных.
