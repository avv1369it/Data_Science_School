# Библиотека Surprise

## 1. Описание библиотеки

<span style="color:blue">**Библиотека Surprise (Simple Python RecommendatIon System Engine)**

**Surprise**— это библиотека Python, предназначенная для построения и оценки алгоритмов рекомендательных систем. Она особенно полезна для тех, кто занимается разработкой систем, основанных на методах коллаборативной фильтрации (Collaborative Filtering). Эта библиотека предоставляет множество готовых алгоритмов, а также инструменты для тестирования и сравнения их производительности.

**Основные возможности Surprise:**

1. ***Алгоритмы коллаборативной фильтрации:*** Surprise включает в себя реализацию популярных алгоритмов коллаборативной фильтрации, таких как:
    * Базовые алгоритмы: KNN (k-ближайших соседей) для пользователей и предметов.
    * SVD (Singular Value Decomposition): Метод разложения матрицы для прогнозирования рейтингов.
    * NMF (Non-negative Matrix Factorization): Неразложимое матричное разложение.
    * SlopeOne: Алгоритм, основанный на различиях между парами элементов.
    * CoClustering: Алгоритм кластеризации, который группирует пользователей и элементы одновременно.

2. ***Оценочные метрики:*** Surprise предоставляет разнообразные метрики для оценки точности прогнозируемых рейтингов, такие как RMSE (Root Mean Square Error), MAE (Mean Absolute Error), FCP (Fraction of Concordant Pairs) и др.

3. ***Кросс-валидация:*** Поддерживает кросс-валидацию для проверки обобщающей способности моделей. Можно легко настроить количество фолдов и выбрать стратегию разделения данных.
4. ***Работа с датасетами:*** Встроена поддержка популярных датасетов, таких как MovieLens и Jester. Также есть возможность загружать собственные данные в формате CSV.
5. ***Параллелизация:*** Для повышения скорости обработки больших объемов данных Surprise поддерживает параллельное выполнение некоторых операций, используя модуль joblib.
6. ***Гибкость и расширяемость:*** Пользователи могут создавать свои собственные алгоритмы и интегрируемые компоненты, что делает библиотеку удобной для исследовательских целей.

#### Функции расчета precision, recall и вывода рекомендаций из дополнительных материалов Surprise 

````python
def precision_recall_at_k(predictions, k=3, threshold=3.5):
    """Функция вычисления метрики precision@k и recall@k
    
    Args:
        predictions (dict): словарь с результатами предсказаний модели.
        k (int, optional): количество рекомендаций / товар по которым необходимо произвести оценку. По умолчанию = 3.
        threshold (float, optional): минимальный уровень достоверности, который должен быть достигнут, 
                                     чтобы элемент считался значимым для расчета метрик точности и полноты. По умолчанию = 0.5

    Returns:
        precisions (float): расчитаная метрика precision@k на выборке
        recalls (float): расчитаная метрика recall@k на выборке
    """
    

    # First map the predictions to each user.
    user_est_true = defaultdict(list)
    for uid, _, true_r, est, _ in predictions:
        user_est_true[uid].append((est, true_r))

    precisions = dict()
    recalls = dict()
    for uid, user_ratings in user_est_true.items():

        # Sort user ratings by estimated value
        user_ratings.sort(key=lambda x: x[0], reverse=True)

        # Number of relevant items
        n_rel = sum((true_r >= threshold) for (_, true_r) in user_ratings)

        # Number of recommended items in top k
        n_rec_k = sum((est >= threshold) for (est, _) in user_ratings[:k])

        # Number of relevant and recommended items in top k
        n_rel_and_rec_k = sum(
            ((true_r >= threshold) and (est >= threshold))
            for (est, true_r) in user_ratings[:k]
        )

        # Precision@K: Proportion of recommended items that are relevant
        # When n_rec_k is 0, Precision is undefined. We here set it to 0.

        precisions[uid] = n_rel_and_rec_k / n_rec_k if n_rec_k != 0 else 0

        # Recall@K: Proportion of relevant items that are recommended
        # When n_rel is 0, Recall is undefined. We here set it to 0.

        recalls[uid] = n_rel_and_rec_k / n_rel if n_rel != 0 else 0
        
        # усредняем Precision и recall по всем пользователям
        precisions_mean = sum(prec for prec in precisions.values()) / len(precisions)
        recalls_mean = sum(rec for rec in recalls.values()) / len(recalls)

    return precisions_mean, recalls_mean
````

````python
def get_top_n(predictions, n=3):
    """Вернуть топ-N рекомендаций для каждого пользователя из набора предсказаний.

    Args:
        predictions (list): Список предсказаний, как возвращено методом test алгоритма.
        n (int): Количество рекомендаций, которое нужно вывести для каждого пользователя. По умолчанию равно 3.

    Returns:
        top_n (dict) - cловарь, где ключи — идентификаторы пользователей (в исходном виде),
            а значения — списки кортежей: [(идентификатор элемента в исходном виде, оценка предсказания), ...] размером n.
    """
    
    # Cопоставим прогнозы каждому пользователю.
    top_n = defaultdict(list)
    for uid, iid, true_r, est, _ in predictions:
        top_n[uid].append((iid, est))

    # Oтсортируем прогнозы для каждого пользователя и извлечем k наивысших.
    for uid, user_ratings in top_n.items():
        user_ratings.sort(key=lambda x: x[1], reverse=True)
        top_n[uid] = user_ratings[:n]
    
    return top_n
````

#### Функции расчета precision, recall и вывода рекомендаций доработанные

````python
# Библиотека Surprise не содержит метрику precision@k, опишем функцию для ее вычисления и оценки работы моделей
def precision_recall_at_k_modified(predictions, k=3, threshold=3.5):
    """Функция вычисления метрики precision@k и recall@k.

    Args:
        predictions (dict): словарь с результатами предсказаний модели.
        k (int, optional): количество рекомендаций / товар по которым необходимо произвести оценку. По умолчанию = 3.
        threshold (float, optional): минимальный уровень достоверности, который должен быть достигнут, 
                                     чтобы элемент считался значимым для расчета метрик точности и полноты. По умолчанию = 0.5

    Returns:
        precision_ (float): расчитаная метрика precision@k на выборке
        recall_ (float): расчитаная метрика recall@k на выборке
    """
    
    from collections import defaultdict # 
    from sklearn.metrics import precision_score # 
    
    # cоздаем словарь, где ключ - user_id, а значение - список из top-k предсказанных элементов
    user_est_true = defaultdict(list)
    for uid, _, true_r, est, _ in predictions:
        user_est_true[uid].append((est, true_r))

    precisions = dict()
    recalls = dict()
    
    # 
    for uid, user_ratings in user_est_true.items():
        # cортируем пользователей
        user_ratings.sort(key=lambda x: x[0], reverse=True)

        n_rel = sum((true_r >= threshold) for (_, true_r) in user_ratings)
        n_rec_k = sum((est >= threshold) for (est, _) in user_ratings[:k])
        
        # Количество релевантных и рекомендуемых товаров в топ k 
        n_rel_and_rec_k = sum(
            ((true_r >= threshold) and (est >= threshold))
            for (est, true_r) in user_ratings[:k]
        )

        # Precision@K: Доля рекомендованных товаров, которые являются актуальными
        # Когда n_rec_k равно 0, Precision не определена. Здесь мы устанавливаем ее равной 0.

        precisions[uid] = n_rel_and_rec_k / n_rec_k if n_rec_k != 0 else 0

        # Recall@K: Доля соответствующих товаров, которые рекомендуются
        # Когда n_rel равно 0, Recall не определен. Здесь мы устанавливем его равным 0.

        recalls[uid] = n_rel_and_rec_k / n_rel if n_rel != 0 else 0

        # усредняем Precision и recall по всем пользователям
        precision_ = sum(prec for prec in precisions.values()) / len(precisions)
        recall_ = sum(rec for rec in recalls.values()) / len(recalls)

    return precision_, recall_
````

#### Дополнительные функции для оценки работы KNN моделей

````python
# опишем вспомогательные функции для оценки работы моделей
def get_model_name(model):
    """Возвращает название модели в виде строки.
    Args:
        model (str): Объект модели, для которой необходимо получить имя.
    Returns:
        str: Строка с названием модели.
    """
    return str(model).split('.')[-1].split(' ')[0].replace("'>", "")

# Генерация словаря моделей
def generate_models_dict(models, sim_names, user_based):
    """Генерирует словарь моделей, где ключом является имя модели, а значением – экземпляр модели.

    Args:
        models (str): Список объектов моделей.
        sim_names (list): Список имен подобий, которые будут использованы для настройки моделей.
        user_based (_type_): Логическая переменная, определяющая, основаны ли модели на пользователях.

    Returns:
        models_dict(dict): Словарь, содержащий экземпляры моделей с ключами, представляющими собой имена моделей и подобий.
    """
    models_dict = {}

    for sim_name in sim_names:
        sim_dict = {
            'name': sim_name,
            'user_based': user_based
        }
        for model in models:
            model_name = get_model_name(model) + ' ' + sim_name
            models_dict[model_name] = model(sim_options=sim_dict)

    return models_dict

# Функция кросс-валидации нескольких моделей
def cv_multiple_models(data, models_dict, cv=3):
    """Выполняет кросс-валидацию для множества моделей, представленных в словаре models_dict, и собирает результаты в датафрейме pandas.

    Args:
        data (dataframe): Набор данных, используемый для кросс-валидации.
        models_dict (dict): Словарь моделей, созданный функцией generate_models_dict.
        cv (int, optional): Количество фолдов для кросс-валидации (по умолчанию 3)..

    Returns:
        results(dataframe): Датафрейм pandas, содержащий средние значения метрик для каждой модели.
    """
    results = pd.DataFrame()

    for model_name, model in models_dict.items():
        print(f'\n---> CV for {model_name}...')

        cv_results = cross_validate(model, data, cv=cv)
        tmp = pd.DataFrame(cv_results).mean().to_frame().T
        tmp['model'] = model_name
        results = pd.concat([results, tmp], ignore_index=True)

    return results
````

## 4. Коллаборативная фильтрация (Collaborative Filtering)

### 4.1. Фильтрация на основе пользователей (User-based CF)

**Пользовательская фильтрация (User-Based)**
>Когда параметр ***user_based=True***, алгоритм ищет пользователей, которые имеют похожие предпочтения с текущим пользователем, и рекомендует ему те предметы, которые понравились этим похожим пользователям. Этот подход основан на том, что если два пользователя оценили одни и те же предметы одинаково, они, вероятно, будут иметь схожий вкус и в будущем.

**<span style="color:green"> Лучшая модель SVDpp. Проведем настройку интервалов гиперпараметров по результатам прошлого подбора и еще раз подбор пораметров с помощью OPTUNA.**

````python
def objective(trial):
    # Получаем гиперпараметры из trial
    n_factors = trial.suggest_int("n_factors", 10, 100)                        # Количество факторов. По умолчанию 20. Оптимальное значение часто находится между 50 и 150.
    cache_ratings = trial.suggest_categorical("cache_ratings", [True, False])  # Кэшировать ли рейтинги во время обучения. По умолчанию False.
    init_mean = trial.suggest_int("init_mean", 0, 1)                           # Среднее значение нормального распределения для инициализации факторных векторов. По умолчанию 0. Обычно достаточно малого значения от 0 до 1.
    init_std_dev = trial.suggest_uniform("init_std_dev", 0.01, 0.1)            # Стандартное отклонение нормального распределения для инициализации факторных векторов. По умолчанию 0.1. Рекомендуется искать в пределах от 0.01 до 0.1.
    lr_all = trial.suggest_loguniform("lr_all", 1e-4, 1e-2)                    # Скорость обучения для всех параметров. По умолчанию 0.007. Логарифмически равномерный поиск между 1e-4 и 1e-2 часто эффективен.
    reg_all = trial.suggest_loguniform("reg_all", 1e-4, 1e-1)                  # Регуляризационный член для всех параметров. По умолчанию 0.02. Логарифмический поиск в диапазоне от 1e-4 до 1e-1 подходит для большинства случаев.
    lr_bu = trial.suggest_loguniform("lr_bu", 1e-4, 1e-1)                      # Скорость обучения для пользователей. Имеет приоритет над lr_all, если установлено. По умолчанию None. Аналогично lr_all, логарифмическое распределение хорошо работает.
    lr_bi = trial.suggest_loguniform("lr_bi", 1e-4, 1e-1)                      # Скорость обучения для элементов. Имеет приоритет над lr_all, если установлено. По умолчанию None. Аналогично lr_all, логарифмическое распределение хорошо работает.
    lr_pu = trial.suggest_loguniform("lr_pu", 1e-4, 1e-1)                      # Скорость обучения для пользовательских факторов. Имеет приоритет над lr_all, если установлено. По умолчанию None. Аналогично lr_all, логарифмическое распределение хорошо работает.
    lr_qi = trial.suggest_loguniform("lr_qi", 1e-4, 1e-1)                      # Скорость обучения для элементных факторов. Имеет приоритет над lr_all, если установлено. По умолчанию None. Аналогично lr_all, логарифмическое распределение хорошо работает.
    lr_yj = trial.suggest_loguniform("lr_yj", 1e-4, 1e-1)                      # Скорость обучения для yj. Имеет приоритет над lr_all, если установлено. По умолчанию None. Аналогично lr_all, логарифмическое распределение хорошо работает.
    reg_bu = trial.suggest_loguniform("reg_bu", 1e-4, 1e-1)                    # Регуляризация для пользователей. Имеет приоритет над reg_all, если установлено. По умолчанию None. Аналогично reg_all, логарифмическое распределение хорошо работает.
    reg_bi = trial.suggest_loguniform("reg_bi", 1e-4, 1e-1)                    # Регуляризация для элементов. Имеет приоритет над reg_all, если установлено. По умолчанию None. Аналогично reg_all, логарифмическое распределение хорошо работает.
    reg_pu = trial.suggest_loguniform("reg_pu", 1e-4, 1e-1)                    # Регуляризация для пользовательских факторов. Имеет приоритет над reg_all, если установлено. По умолчанию None. Аналогично reg_all, логарифмическое распределение хорошо работает.
    reg_qi = trial.suggest_loguniform("reg_qi", 1e-4, 1e-1)                    # Регуляризация для элементных факторов. Имеет приоритет над reg_all, если установлено. По умолчанию None. Аналогично reg_all, логарифмическое распределение хорошо работает.
    reg_yj = trial.suggest_loguniform("reg_yj", 1e-4, 1e-1)                    # Регуляризация для yj. Имеет приоритет над reg_all, если установлено. По умолчанию None. Аналогично reg_all, логарифмическое распределение хорошо работает.

    # Обучение модели с новыми параметрами
    model = SVDpp(
        n_factors=n_factors,
        n_epochs=50,                                                          # Количество итераций процедуры SGD. По умолчанию 20
        cache_ratings=cache_ratings,
        init_mean=init_mean,
        init_std_dev=init_std_dev,
        lr_all=lr_all,
        reg_all=reg_all,
        lr_bu=lr_bu,
        lr_bi=lr_bi,
        lr_pu=lr_pu,
        lr_qi=lr_qi,
        lr_yj=lr_yj,
        reg_bu=reg_bu,
        reg_bi=reg_bi,
        reg_pu=reg_pu,
        reg_qi=reg_qi,
        reg_yj=reg_yj,
        random_state=RANDOM_SEED,
        verbose=False,
    )

    model.fit(trainset)

    # Оценка модели
    predictions = model.test(testset)
    metric_rmse = rmse(predictions)

    return metric_rmse

study = optuna.create_study(direction="minimize")
study.optimize(objective, n_trials=20, show_progress_bar=True)

# выводим наилучшее достигнутое значение метрики
best_metric = study.best_value
print(f'Наилучшая достигнутая метрика RMSE: {best_metric}')
# выводим параметры при которых достигнута митрика
best_params = study.best_trial.params
print(f'Наилучшие параметры: {best_params}')
````
