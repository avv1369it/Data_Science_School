## 1. Описание библиотеки

**Recommenders** – это библиотека Python для создания рекомендательных систем. Она была разработана компанией RecSys Ventures, специализирующейся на разработке решений для рекомендаций контента пользователям. Библиотека предоставляет набор инструментов и алгоритмов, которые помогают создавать персонализированные рекомендации на основе предпочтений пользователей, их поведения и других факторов.

**Основные возможности библиотеки Recommenders:**
1. **Коллаборативная фильтрация:** Один из самых популярных методов построения рекомендаций. Он основан на анализе пользовательских оценок и предпочтений, чтобы предсказать интерес к новым элементам. В библиотеке реализованы такие методы коллаборативной фильтрации, как SVD (Singular Value Decomposition), NMF (Non-negative Matrix Factorization) и другие.
2. **Контентная фильтрация:** Этот метод основывается на анализе характеристик самих элементов (например, фильмов, книг, товаров). Алгоритмы контентной фильтрации используют информацию об элементах, чтобы находить похожие элементы для каждого пользователя.
3. **Гибридные подходы:** Комбинация различных методов фильтрации позволяет создать более точные рекомендации. Например, можно объединить коллаборативную фильтрацию с контентной для достижения лучших результатов.
4. **Оценка качества рекомендаций:** Библиотека включает инструменты для оценки точности и релевантности предложенных рекомендаций. Это помогает оптимизировать модели и улучшать качество работы системы.
5. **Поддержка различных источников данных:** Recomenders поддерживает работу с различными форматами данных, включая CSV-файлы, базы данных SQL и другие источники.
6. **Визуализация и анализ данных:** В состав библиотеки входят инструменты для визуализации и анализа данных, что облегчает процесс разработки и тестирования моделей.
Интеграция с другими библиотеками: Recomenders легко интегрируется с популярными библиотеками машинного обучения, такими как Scikit-learn, TensorFlow и PyTorch, что расширяет возможности её использования.

````python
# cоздаем пустой DataFrame для хранения результатов
results = pd.DataFrame(columns=["Model", "Time", "train_Precision@3", "test_Precision@3"])

def add_results(dataframe, model_name, elapsed_time, train_precision, test_precision):
    """Добавляет результаты работы модели в датафрейм.

    Args:
        dataframe (DataFrame): Датафрейм, содержащий результаты предыдущих моделей.
        model_name (str): Название текущей модели.
        elapsed_time (float): Время, затраченное на обучение и тестирование модели.
        train_precision (float): Значение метрики Precision@3 на обучающей выборке.
        test_precision (float): Значение метрики Precision@3 на тестовой выборке.
    
    Returns:
        dataframe (DataFrame): Обновленный датафрейм с добавленными результатами текущей модели.
    """
    new_row = pd.Series({
        "Model": model_name,
        "Time": elapsed_time,
        "train_Precision@3": train_precision,
        "test_Precision@3": test_precision
    })
    dataframe = pd.concat([dataframe, new_row.to_frame().T], ignore_index=True)
    return dataframe
````

## 3. Коллаборативная фильтрация (Collaborative Filtering)

### 3.1 Построим модель на основе алгоритма Bilateral Variational Autoencoder (BiVAE) 

**Модели двустороннего вариационного автокодировщика (BiVAE)**, применяемого для коллаборативной фильтрации. Исследование, посвященное модели BiVAE [1], было представлено на конференции WSDM'21. Все эксперименты, связанные с моделью BiVAE, описаны в этом [репозитории](https://github.com/PreferredAI/bi-vae).

Реализация модели взята из [Cornac](https://github.com/PreferredAI/cornac) [2], который является фреймворком для мультимодальных рекомендательных систем, ориентированных на модели, использующие вспомогательные данные (например, описательный текст и изображения товаров, социальные сети и т.д.).

BiVAE - это новый вариационный автоэнкодер, разработанный специально для диадических данных, где наблюдения состоят из измерений, связанных с двумя наборами объектов, например, пользователями, предметами и соответствующими рейтингами. Модель симметрична, что упрощает расширение вспомогательных данных с обеих сторон от пользователей и элементов. В дополнение к данным предпочтений, модель может быть применена к другим типам двоичных данных, таким как матрицы документ-слово, и другим задачам, таким как совместная кластеризация.

В статье также обсуждается ограниченная адаптивная априорность (CAP), предлагаемый метод построения информативных априорных значений для смягчения хорошо известной проблемы последующего коллапса. Мы намеренно опустили эту часть, чтобы не отвлекать аудиторию. Тем не менее, это очень интересно, и на это стоит взглянуть.

**Описание параметров инициализации модели `cornac.models.BiVAECF`:**

1. **k** (`LATENT_DIM`) – Размер скрытого пространства. Это количество латентных факторов, которое будет использоваться для представления пользователей и элементов в пространстве пониженной размерности.
2. **encoder_structure** (`ENCODER_DIMS`) – Список целых чисел, определяющий структуру слоев энкодера. Каждый элемент списка указывает на количество нейронов в соответствующем слое. Например, `[1000, 500, 200]` означает, что у вас есть три слоя с 1000, 500 и 200 нейронами соответственно.
3. **act_fn** (`ACT_FUNC`) – Функция активации, которая используется между слоями нейронной сети. Примеры функций активации включают ReLU, Sigmoid, Tanh и другие.
4. **likelihood** (`LIKELIHOOD`) – Тип вероятностной модели, используемый для декодирования. Возможными значениями могут быть, например, `'gaussian'`, `'bernoulli'`.
5. **n_epochs** (`NUM_EPOCHS`) – Общее количество эпох (циклов) обучения модели.
6. **batch_size** (`BATCH_SIZE`) – Размер мини-батчей, используемых при обучении. Это количество образцов данных, обрабатываемых одновременно перед обновлением весов модели.
7. **learning_rate** (`LEARNING_RATE`) – Скорость обучения, которая определяет величину шага обновления весов модели после каждой итерации.
8. **seed** (`SEED`) – Начальное значение случайного числа, обеспечивающее воспроизводимость результатов экспериментов.
9. **use_gpu** – Логическое значение, указывающее, использовать ли GPU для ускорения вычислений. Здесь значение устанавливается автоматически в зависимости от доступности CUDA на устройстве.
10. **verbose** – Если установлено в `True`, то выводится информация о ходе процесса обучения (например, прогресс бар).

````python
# Определяем переменные
# Количество рекомендаций для каждого пользователя
TOP_K = 3
# Параметры модели
LATENT_DIM = 50
ENCODER_DIMS = [100]
ACT_FUNC = "tanh"
LIKELIHOOD = "pois"
NUM_EPOCHS = 10
BATCH_SIZE = 256
LEARNING_RATE = 0.001
````

**Cornac Dataset**

Для работы с моделями, реализованными в Cornac, нам нужно создать объект класса [Dataset](https://cornac.readthedocs.io/en/latest/data.html#module-cornac.data.dataset).

Класс Dataset в Cornac служит основным объектом, с которым будут взаимодействовать модели. Помимо преобразования данных, класс Dataset предоставляет множество полезных итераторов для перебора данных, а также поддерживает различные методы выборки отрицательных примеров.

````python
# Создаем Dataset Cornac
train_set = cornac.data.Dataset.from_uir(train.itertuples(index=False), seed=SEED)
print("Количество клиентов: {}".format(train_set.num_users))
print("Количество товаров: {}".format(train_set.num_items))
````

````python
# Определяем параметры модели и обучаем ее
bivae = cornac.models.BiVAECF(
    k=LATENT_DIM,                        # Размер скрытого пространства (количество латентных факторов)
    encoder_structure=ENCODER_DIMS,      # Структура слоев энкодера (список размеров скрытых слоев)
    act_fn=ACT_FUNC,                     # Функция активации для нейронной сети
    likelihood=LIKELIHOOD,               # Вероятностная модель (например, 'gaussian', 'bernoulli')
    n_epochs=NUM_EPOCHS,                 # Количество эпох обучения
    batch_size=BATCH_SIZE,               # Размер батча при обучении
    learning_rate=LEARNING_RATE,         # Скорость обучения
    seed=SEED,                           # Случайное зерно для воспроизводимости результатов
    use_gpu=torch.cuda.is_available(),   # Использовать GPU, если доступен
    verbose=True                         # Выводить информацию о процессе обучения
)

with Timer() as t:
    bivae.fit(train_set)
print("Took {} seconds for training.".format(t))
````

**Prediction and Evaluation**

Теперь, когда наша модель обучена, мы можем создавать ранжированные списки для рекомендаций. Все рекомендательные модели в Cornac предоставляют методы rate () и rank () для прогнозирования номинальной стоимости товара, а также ранжированного списка товаров для данного пользователя. Чтобы использовать текущие схемы оценки, мы будем использовать функции predict() и predict_ranking() внутри cornac_utils для получения прогнозов.

Давайте измерим рекомендательную эффективность модели, используя показатели рейтинга top-K.
