# Библиотека Implicit

## 1. Описание библиотеки

**Implicit** — это библиотека Python, предназначенная для работы с неявными (implicit) предпочтениями пользователей в рекомендательных системах. Разработанная Бенджамином Картером (Benjamin Carter) и поддерживаемая активным сообществом, библиотека фокусируется на инструментах для работы с методами факторизационных машин (factorization machines) и другими алгоритмами, используемыми для обработки неявных предпочтений.

Основное назначение библиотеки Implicit заключается в создании рекомендательных систем, основанных на факторизации матриц. Она особенно полезна для задач коллаборативной фильтрации, таких как рекомендации товаров пользователям на основе их предыдущих взаимодействий с этими товарами.

### Основные особенности библиотеки Implicit:

1. **Работа с неявными предпочтениями:**
   - В отличие от явных предпочтений (explicit preferences), когда пользователи прямо выражают свою оценку объектов (например, ставя звездочки фильмам), неявные предпочтения подразумеваются косвенно через поведение пользователя (например, клики, просмотры страниц, добавление в корзину). Implicit специализируется именно на обработке таких данных.

2. **Реализация алгоритмов факторизационных машин - Factorization Machines (FM):**
    - **Alternating Least Squares (ALS)**: Алгоритм чередующихся наименьших квадратов, который хорошо подходит для задач с плотными матрицами взаимодействия пользователей и предметов.
    - **Bayesian Personalized Ranking (BPR)**: Байесовский персонализированный ранжирующий алгоритм, ориентированный на максимизацию вероятности того, что пользователь выберет один объект перед другим.
    - **Logistic Matrix Factorization (LMF)**: Логистическая факторизация матриц, которая использует логистическую функцию для минимизации ошибки прогнозирования.

3. **Реализация алгоритмов коллаборативной фильтрации item-based (Алгоритмы k-ближайших соседей (KNN)):**
   - **CosineRecommender**: реализует алгоритм ближайших соседей на основе косинусного расстояния между элементами. Косинусное расстояние измеряет угол между двумя векторами, представляющими элементы, и является популярным методом для определения подобия в векторных пространствах. Эта модель подходит для случаев, когда необходимо учитывать направление векторов, а не их длину.
    - **TFIDFRecommender**: строит рекомендации на основе TF-IDF (Term Frequency-Inverse Document Frequency) расстояний между элементами. TF-IDF взвешивает термины таким образом, что редкие термины получают больший вес, чем частые. Этот метод полезен, когда нужно учесть важность редких событий или взаимодействий.
    - **BM25Recommender**: использует метрику BM25 (Best Match 25) для расчета подобия между элементами. BM25 является улучшенной версией TF-IDF, которая учитывает длину документа и частоту термина в документе. Этот метод эффективен для поиска релевантных документов в больших коллекциях текстов.

4. **Оптимизированная производительность:**
   - Для ускорения вычислений критически важные части кода написаны на Cython, что существенно повышает производительность библиотеки при работе с большими объемами данных.
   - Поддерживается многопоточное выполнение операций, что позволяет эффективно использовать ресурсы многоядерных процессоров.

5. **Удобство использования:**
   - Implicit предоставляет простой и интуитивно понятный API, который облегчает подготовку данных, обучение моделей и получение рекомендаций. Это делает библиотеку удобной для интеграции в существующие проекты.

6. **Открытый исходный код:**
   - Исходный код библиотеки доступен на GitHub, что дает возможность пользователям вносить изменения и улучшения, а также участвовать в развитии проекта. Открытая экосистема способствует активному обсуждению и улучшению функционала библиотеки.

### Преимущества использования Implicit:

- **Эффективность:** библиотека оптимизирована для работы с большими массивами данных, что делает ее подходящей для крупных проектов.
- **Широкий спектр алгоритмов:** наличие нескольких алгоритмов позволяет выбрать наиболее подходящий для конкретных задач и данных.
- **Совместимость:** Implicit легко интегрируется с популярными фреймворками для машинного обучения, такими как Scikit-learn и TensorFlow.
- **Активное сообщество:** постоянная поддержка и развитие библиотеки благодаря активному сообществу разработчиков и пользователей.

## 2. Построение моделей на базе алгоритмов "Коллаборативной фильтрация (англ. Collaborative Filtering)" 

**Коллаборативная фильтрация (англ. Collaborative Filtering)** — это одна из ключевых техник, применяемых в рекомендательных системах для предсказания предпочтений пользователей на основе их прошлых взаимодействий с объектами (например, покупки товаров, просмотр фильмов, прослушивание музыки и т.д.). Основная идея заключается в том, что пользователи с похожими вкусами или поведением склонны выбирать одинаковые объекты. Таким образом, система может рекомендовать новые объекты одному пользователю на основании того, что эти объекты понравились другим пользователям с аналогичными предпочтениями.

### Принципы работы коллаборативной фильтрации:

1. **Сбор данных о взаимодействии пользователей и объектов:**
   - Система собирает информацию о том, какие объекты (фильмы, книги, товары и т.д.) пользователи оценили, купили, посмотрели или добавили в избранное. Это могут быть как явные оценки (рейтинги), так и неявные признаки интереса (просмотры, клики, добавления в корзину).

2. **Определение схожести между пользователями или объектами:**
   - На основе собранных данных система вычисляет степень схожести между пользователями или объектами. Например, если два пользователя оценили одинаково несколько фильмов, считается, что их вкусы похожи.

3. **Предсказание предпочтений:**
   - Когда системе нужно сделать рекомендацию новому пользователю, она ищет пользователей с похожими предпочтениями и предлагает ему объекты, которые понравились этим пользователям, но ещё не были оценены новым пользователем.

### Типы коллаборативной фильтрации:

1. **User-based (основанный на пользователях):**
   - Этот подход ищет пользователей с похожими вкусами и рекомендует объекты, которые понравились этим пользователям. Пример: если Анна и Борис любят одни и те же фильмы, системе стоит порекомендовать Анне фильмы, которые Борис смотрел, но Анна ещё нет.

2. **Item-based (основанный на объектах):**
   - Этот подход ищет объекты, подобные тем, которые пользователь уже оценил положительно, и рекомендует их. Пример: если Анна любит фильмы жанра комедии, системе следует предложить ей другие комедии, которые она ещё не смотрела.

### Преимущества коллаборативной фильтрации:

- **Универсальность:** Может применяться в различных сферах, таких как электронная коммерция, музыкальные сервисы, видеостриминг и социальные сети.
- **Простота внедрения:** Не требует глубокого понимания предметной области, достаточно иметь данные о взаимодействиях пользователей и объектов.
- **Высокое качество рекомендаций:** Часто даёт точные и релевантные рекомендации, особенно в условиях большого количества данных.

### Недостатки коллаборативной фильтрации:

- **Холодный старт:** Трудно делать рекомендации новым пользователям или объектам, пока не накоплено достаточное количество данных о них.
- **Проблема спама:** Могут возникнуть сложности с фильтрацией ложных рейтингов или отзывов, что может исказить результаты.
- **Разреженность данных:** Матрица взаимодействий пользователей и объектов часто бывает очень разреженной, что усложняет обработку данных.

### Примеры использования:

- **Amazon:** Рекомендует товары на основе покупок и просмотров других пользователей.
- **Netflix:** Предлагает фильмы и сериалы на основе того, что смотрели и оценивали другие зрители.
- **Spotify:** Подбирает плейлисты и треки на основе музыкальных предпочтений пользователей.

Коллаборативная фильтрация остаётся одной из самых популярных и эффективных технологий для создания персональных рекомендаций, несмотря на свои недостатки.

### 2.1. Модель "NNM CosineRecomender"

**Модель `CosineRecommender`** из библиотеки Implicit использует косинусное сходство для расчета рекомендаций. Косинусное сходство измеряет угол между двумя векторами, что позволяет оценить схожесть между пользователями или элементами на основе их предпочтений. Рассмотрим ключевые параметры этой модели и предложим разумные диапазоны значений для них.

#### Основные параметры модели CosineRecommender:

1. **K**: Количество ближайших соседей, которые используются для формирования рекомендаций. Этот параметр влияет на точность и разнообразие рекомендаций. Чем больше значение K, тем больше элементов будет рассматриваться для формирования рекомендаций, однако это может увеличить вычислительные затраты. **Рекомендуемые значения:** 5–50.

2. **num_threads**: Количество потоков, используемых для параллельных вычислений. Этот параметр важен для увеличения скорости обработки больших наборов данных.

````python
# фиксируем время начала работы
start_time = time.time()
# формируем параметры модели и обучаем её
model = CosineRecommender(
    K=16,             # стандартное количество ближайших соседей
    num_threads=8     # количество ядер
)

model.fit(train)

# расчитываем метрику precision@k
p_at_k = precision_at_k(model, train, test, K=3)
p_at_k_train = 0
# фиксируем время окончания работы
end_time = time.time()
elapsed_time = end_time - start_time

print(f'Precision@3: {p_at_k:.4f}')
print()

# Вариан 1 - Выбираем случайного пользователя
random_normalized_visitorid = np.random.choice(users)

# Вариант 2 - Вводим оригинальный visitorid
# original_visitorid = 391657
# random_normalized_visitorid = original_visitorid - min_visitorid # приводим его к нормализованному виду

userid = user_to_idx[random_normalized_visitorid]

# получаем рекомендации для пользователя
ids, scores = model.recommend(userid, matrix[userid], N=3, filter_already_liked_items=False)
# преобразуем обратно в оригинальные itemid
original_item_ids = [items[idx] + min_itemid for idx in ids]
# преобразуем обратно в оригинальный visitorid
original_visitorid = random_normalized_visitorid + min_visitorid

# используем pandas для отображения результатов в виде таблицы
rec_user = pd.DataFrame({
    "visitorid": original_visitorid,
    "itemid": original_item_ids,
    "score": scores,
    "already_liked": np.in1d(ids, matrix[userid].indices.ravel())  # преобразуем в одномерный массив
})
print(rec_user)
print()
set1 = set(df[df["visitorid"] == original_visitorid]["itemid"].unique().tolist())
set2 = set(original_item_ids)
print(f'Соответсвие itemid между рекомендациями и фактом {set1&set2}')
print()
print(rec_user["already_liked"].value_counts())

# сохраняем результаты
results_implicit = add_results(results_implicit, "NNM_CosinRecomender", elapsed_time, p_at_k_train, p_at_k)
display(results_implicit)
````

### Подбор гиперпараметров с помощью **OPTUNA**

### 2.3. Модель `BM25Recommender`

**Модель `BM25Recommender`** из библиотеки Implicit использует алгоритм BM25 (Best Match 25) для ранжирования элементов на основе их релевантности пользователям. Этот алгоритм изначально использовался в информационных поисковых системах, но оказался эффективным и в рекомендательных системах. Рассмотрим ключевые параметры этой модели и предложим разумные диапазоны значений для них.

#### Основные параметры модели BM25Recommender:
1. **K**: Количество ближайших соседей, которые используются для формирования рекомендаций. Этот параметр влияет на точность и разнообразие рекомендаций. 2. **K1**: Параметр, контролирующий степень сглаживания частотности термина. Чем выше значение K1, тем больше влияние имеет частота взаимодействий. **Рекомендуемые значения:** 1.2–2.0.

4. **num_threads**: Количество потоков, используемых для параллельных вычислений. Этот параметр важен для увеличения скорости обработки больших наборов данных. **Рекомендуемые значения:** зависит от количества ядер процессора, обычно 1–8.

## 3. Построение моделей на базе алгоритмов "Матричной факторизации и факторизационных машин."

***Матричная факторизация и факторизационные машины*** — это мощные инструменты, которые используются в рекомендательных системах (РС) для предсказания предпочтений пользователей на основе их прошлого поведения. Давайте разберём оба этих понятия отдельно.

### Матричная факторизация (Matrix Factorization)

**Матричная факторизация** — это метод, который пытается найти скрытую структуру в данных, представленных в виде матрицы. В контексте рекомендательных систем, эта матрица обычно содержит рейтинги или взаимодействия пользователей с объектами (например, фильмы, песни, товары). Цель матричной факторизации — представить эту большую и зачастую разреженную матрицу в виде произведения двух меньших матриц, которые содержат скрытые факторы (латентные переменные).

#### Как это работает?

Предположим, у нас есть матрица $R$ размером $m \times n$, где $m$ — количество пользователей, а $n$ — количество объектов. Мы хотим разложить эту матрицу на два фактора: матрицу $P$ размером $m \times k$ и матрицу $Q$ размером $k \times n$. Здесь $k$ — это количество скрытых факторов, которые мы выбираем заранее.

Процесс выглядит так:

$$ R \approx P \cdot Q^T

$$

Где $Q^T$ — транспонированная матрица $Q$.

#### Что означают эти скрытые факторы?

Скрытые факторы представляют собой абстрактные характеристики, которые объясняют предпочтения пользователей и свойства объектов. Например, в случае рекомендательной системы фильмов, эти факторы могут включать жанры, актёров, режиссеров и т.д. Пользователи и объекты могут быть описаны в терминах этих факторов, что позволяет системе находить соответствия между ними.

#### Почему это полезно?

Матричная факторизация позволяет уменьшить размерность данных, что делает расчёты быстрее и эффективнее. Кроме того, она помогает справиться с проблемой разреженности данных, так как заполняет пропущенные значения в матрице рейтингов на основе найденных скрытых факторов.

### Факторизационные машины (Factorization Machines)

**Факторизационные машины (FM)** — это расширение идеи матричной факторизации, которое было предложено Штейнером и Фридманом в 2010 году. В отличие от обычной матричной факторизации, которая рассматривает только пары взаимодействий (пользователь-объект), факторизационные машины также учитывают взаимодействия между различными признаками.

#### Как это работает?

Факторизационная машина принимает на вход вектор признаков $x$ и вычисляет предсказанное значение $y$ следующим образом:

$$ \hat{y}(x) = w_0 + \sum_{i=1}^n w_i x_i + \sum_{i=1}^n \sum_{j=i+1}^n \langle v_i, v_j \rangle x_i x_j

$$

Здесь:
- $w_0$ — глобальный сдвиг,
- $w_i$ — вес признака $i$,
- $v_i$ и $v_j$ — векторы скрытых факторов для признаков $i$ и $j$,
- $\langle v_i, v_j \rangle$ — скалярное произведение векторов $v_i$ и $v_j$.

#### Зачем нужны взаимодействия признаков?

Взаимодействия признаков позволяют учитывать более сложные зависимости между разными аспектами данных. Например, в электронной коммерции взаимодействие между характеристиками товара и историей покупок пользователя может дать более точную картину его предпочтений, чем просто сумма отдельных характеристик.

#### Преимущества факторизационных машин:

- Учитываются взаимодействия между признаками, что приводит к более точным моделям.
- Хорошо справляются с разреженными данными, что делает их полезными в рекомендательных системах.
- Универсальны и могут применяться не только для задач классификации и регрессии, но и для задач ранжирования.

### Сравнение матричной факторизации и факторизационных машин:

- **Матричная факторизация** фокусируется на разбиении большой матрицы взаимодействий на две меньшие матрицы, содержащие скрытые факторы. Она проста в понимании и реализации, но ограничена парными взаимодействиями.
  
- **Факторизационные машины** идут дальше, учитывая взаимодействия между различными признаками, что делает их более мощными и гибкими. Они сложнее в реализации, но предлагают лучшую точность и универсальность.

Оба метода играют важную роль в рекомендательных системах, помогая создавать персонализированные рекомендации на основе истории взаимодействий пользователей с объектами.

### 3.1. Модель `AlternatingLeastSquares` (ALS)

**Модель `AlternatingLeastSquares` (ALS)** — это еще один популярный алгоритм для построения рекомендательных систем, основанная на факторизационных методах. Она широко применяется благодаря своей эффективности и возможности масштабироваться на большие наборы данных. Рассмотрим ключевые параметры этой модели и предложим разумные диапазоны значений для них.

#### Основные параметры модели ALS:

1. **factors**: Количество скрытых факторов, используемых для представления пользователей и товаров. Этот параметр влияет на сложность модели. Обычно его выбирают в диапазоне от 50 до 300, хотя иногда используют и большие значения, если требуется высокая точность. **Рекомендуемые значения:** 50–200
2. **regularization**: Регуляризация предотвращает переобучение модели, добавляя штраф за сложность параметров. Высокие значения регуляризации могут ухудшить способность модели обобщать новые данные, низкие — приведут к переобучению. **Рекомендуемые значения:** 0.01–0.1
3. **alpha**: Этот параметр используется для вычисления взвешенных оценок. Он добавляет дополнительный вес позитивным примерам (например, купленным товарам) по сравнению с негативными (некупленными). **Рекомендуемые значения:** 1–40
4. **iterations**: Количество эпох обучения. Определяет, сколько раз модель пройдет через весь набор данных. Обычно выбирается в зависимости от размера набора данных и сложности задачи. **Рекомендуемые значения:** 10–30
5. **use_cg**: Использование метода сопряжённых градиентов для оптимизации. Может ускорить процесс обучения, но требует больше памяти. **Рекомендуемые значения:** True или False
6. **calculate_training_loss**: Включает расчет потерь на этапе обучения. Полезно для мониторинга процесса обучения. **Рекомендуемые значения:** True или False

````python
# фиксируем время начала работы
start_time = time.time()
# формируем параметры модели и обучаем её
model = AlternatingLeastSquares(
    factors=100,                   # Увеличим количество факторов для повышения точности
    regularization=0.01,           # Среднее значение регуляризации
    alpha=1,                       # Добавляем умеренное смещение к позитивным примерам
    iterations=15,                 # Разумное количество итераций
    use_native=True,
    use_cg=True,                   # Используем метод сопряжённых градиентов
    calculate_training_loss=False, # Рассчитываем потери на этапе обучения
    random_state=RANDOM_SEED,
    num_threads=8
)

model.fit(train)

# расчитываем метрику precision@k
p_at_k = precision_at_k(model, train, test, K=3)
p_at_k_train = 0
# фиксируем время окончания работы
end_time = time.time()
elapsed_time = end_time - start_time

print(f'Precision@3: {p_at_k:.4f}')
print()
# Вариан 1 - Выбираем случайного пользователя
random_normalized_visitorid = np.random.choice(users)

# Вариант 2 - Вводим оригинальный visitorid
# original_visitorid = 391657
# random_normalized_visitorid = original_visitorid - min_visitorid # приводим его к нормализованному виду

userid = user_to_idx[random_normalized_visitorid]

# получаем рекомендации для пользователя
ids, scores = model.recommend(userid, matrix[userid], N=3, filter_already_liked_items=False)
# преобразуем обратно в оригинальные itemid
original_item_ids = [items[idx] + min_itemid for idx in ids]
# преобразуем обратно в оригинальный visitorid
original_visitorid = random_normalized_visitorid + min_visitorid

# используем pandas для отображения результатов в виде таблицы
rec_user = pd.DataFrame({
    "visitorid": original_visitorid,
    "itemid": original_item_ids,
    "score": scores,
    "already_liked": np.in1d(ids, matrix[userid].indices.ravel())  # преобразуем в одномерный массив
})
print(rec_user)
print()
set1 = set(df[df["visitorid"] == original_visitorid]["itemid"].unique().tolist())
set2 = set(original_item_ids)
print(f'Соответсвие itemid между рекомендациями и фактом {set1&set2}')
print()
print(rec_user["already_liked"].value_counts())
print()

# сохраняем результаты
results_implicit = add_results(results_implicit, "MF_ALS", elapsed_time, p_at_k_train, p_at_k)
display(results_implicit)
````

````python
def objective(trial):
    # Получаем гиперпараметры из trial
    factors = trial.suggest_int("factors", 50, 200)
    regularization = trial.suggest_loguniform("regularization", 1e-4, 1e-1)
    alpha = trial.suggest_uniform("alpha", 0.5, 10)
    use_native = trial.suggest_categorical("use_native", ["False", "True"])
    use_cg = trial.suggest_categorical("use_cg", ["False", "True"])
    calculate_traning_loss = trial.suggest_categorical("calculate_training_loss", ["False", "True"])
    
    # Формируем и обучаем модель с новыми параметрами
    model = AlternatingLeastSquares(
        factors=factors,
        regularization=regularization,
        alpha=alpha,
        iterations=20,
        use_native=use_native,
        use_cg=use_cg,
        calculate_training_loss=calculate_traning_loss,
        random_state=RANDOM_SEED,
        num_threads=8
    )
    model.fit(train)

    # Оцениваем модель
    p_at_k = precision_at_k(model, train, test, K=3)
    
    return p_at_k

study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=25, show_progress_bar=True)

best_params = study.best_trial.params
print("Лучшие параметры:", best_params)
````

### 4. Модель `NMSLibModel` - <span style="color:red"> с данной надстройкой и библиотекой nmslib до конца не разобралcя 

**Модель `NMSLibModel`** из библиотеки Implicit представляет собой оболочку для использования библиотеки Non-Metric Space Library (NMSLIB) внутри Implicit. NMSLIB — это библиотека для эффективного поиска ближайшего соседа в метрических и неметрических пространствах. Модель `NMSLibModel` позволяет использовать мощные алгоритмы NMSLIB для рекомендательных систем.

#### Основные особенности модели `NMSLibModel`:

1. **Эффективный поиск ближайшего соседа**: NMSLIB предоставляет высокоэффективные алгоритмы для поиска ближайших соседей даже в больших наборах данных. Это особенно полезно для задач, связанных с коллаборативной фильтрацией, где требуется находить похожие элементы или пользователей.

2. **Поддержка различных метрик**: NMSLIB поддерживает множество различных метрик расстояния, включая евклидово расстояние, косинусное сходство, манхэттенское расстояние и многие другие. Это позволяет выбирать подходящую метрику в зависимости от специфики задачи.

3. **Параллельная обработка**: NMSLIB поддерживает параллельную обработку данных, что ускоряет работу с большими объемами информации.

4. **Простота интеграции**: Модель `NMSLibModel` легко интегрируется с библиотекой Implicit, позволяя использовать преимущества NMSLIB без необходимости глубокого погружения в детали этой библиотеки.

#### Пример использования `NMSLibModel`:

```python
from implicit.nearest_neighbours import NMSLibModel

# Создание модели
model = NMSLibModel(method='hnsw', space='cosinesimil', index_params={'M': 16, 'post': 0}, query_time_params={})

# Обучение модели
model.fit(train)

# Получение рекомендаций
user_id = 123
n_recommendations = 10
recommendations = model.recommend(user_id, n_recommendations)
```

#### Ключевые параметры `NMSLibModel`:

1. **method**: Метод индексирования, поддерживаемый NMSLIB. Возможные значения включают `hnsw`, `sw-graph`, `vp-tree` и другие.

2. **space**: Пространство, в котором выполняется поиск. Поддерживаются различные метрики, такие как `l2` (евклидово расстояние), `cosinesimil` (косинусное сходство) и другие.

3. **index_params**: Параметры индексации, зависящие от выбранного метода. Например, для метода `hnsw` параметр `M` определяет количество соседей для каждого элемента.

4. **query_time_params**: Параметры времени запроса, влияющие на точность и скорость поиска.

#### Заключение

Модель `NMSLibModel` предоставляет мощный инструмент для поиска ближайших соседей в рамках рекомендательных систем. Благодаря поддержке различных метрик и высокопроизводительных алгоритмов, она может существенно улучшить качество рекомендаций, особенно в условиях больших объемов данных.
